```python
#Print all elements of a tuple.
def printele(tpl):
    for i in tpl:
        print(i)
tpl=(10, 20, 30, 40, 50)
printele(tpl)
```

```python
#Find the length without using len().
def length(tpl):
    count=0
    for i in tpl:
        count+=1
    return count
tpl=(5, 10, 15, 20)
length(tpl)
```

```python
#Find the sum without using sum().
def sums(tpl):
    sum=0
    for i in tpl:
        sum+=i
    return sum
tpl=(2, 4, 6, 8, 10)
sums(tpl)
```

```python
#Find the largest element without max().
def largest(tpl):
    large=tpl[0]
    for i in range (len(tpl)):
        if large<tpl[i]:
            large=tpl[i]
    return large
tpl=tuple(map(int,input().split()))
largest(tpl)
            
```

```python
#Find the smallest element without min()
def smallest(tpl):
    small=tpl[0]
    for i in range(len(tpl)):
        if small>tpl[i]:
            small=tpl[i]
    return small
tpl=tuple(map(int,input().split()))
smallest(tpl)
```

```python
#Find the average of tuple elements.
def average(tpl):
    sum=0
    for i in range (len(tpl)):
        sum+=tpl[i]
    avg=sum/len(tpl)
    return avg
tpl=tuple(map(int,input().split()))
average(tpl)
```

```python
#Count positive, negative and zero elements.
def count(tpl):
    pos=0
    neg=0
    zero=0
    for i in tpl:
        if i>0:
            pos+=1
        elif i==0:
            zero+=1
        else:
            neg+=1
    return pos,neg,zero
tpl=tuple(map(int,input().split()))
count(tpl)
```

```python
#Search for a given element.
def search(tpl,element):
    for i in tpl:
        if element==i:
            return "Found"
    return "Not found"
tpl=tuple(map(int,input().split()))
element=int(input("element:"))
search(tpl,element)
```

```python
#Count occurrences of an element.
def occur(tpl,element):
    count=0
    for i in tpl:
        if element==i:
            count+=1
    return count
tpl=tuple(map(int,input().split()))
element=int(input("element:"))
occur(tpl,element)
```

```python
#Find the first occurrence.
def firstoccur(tpl,element):
    for i in range (len(tpl)):
        if element==tpl[i]:
            break
    return i
tpl=tuple(map(int,input().split()))
element=int(input("element:"))
firstoccur(tpl,element)
            
```

```python
#find the last occurence:
def lastoccur(tpl,element):
    last=0
    for i in range(len(tpl)):
        if element==tpl[i]:
            last=i
    return last
tpl=tuple(map(int,input().split()))
element=int(input("element:"))
lastoccur(tpl,element)
```

```python
#Print all indexes of an element.
def indexes(tpl,element):
    for i in range (len(tpl)):
        if element==tpl[i]:
            print(i)
tpl=tuple(map(int,input().split()))
element=int(input("element:"))
indexes(tpl,element)
```

```python
#Create a new tuple containing only even elements.
def even(tpl):
    new=[]
    for i in tpl:
        if i%2==0:
            new.append(i)
    return tuple(new)
tpl=tuple(map(int,input().split()))
even(tpl)
```

```python
#Create a new tuple containing squares.
def squares(tpl):
    new=[]
    for i in tpl:
        new.append(i**2)
    return tuple(new)
tpl=tuple(map(int,input().split()))
squares(tpl)
```

```python
#Check whether a tuple is a palindrome.
def palin(tpl):
    i=0
    j=len(tpl)-1
    while i<j:
        if tpl[i]!=tpl[j]:
            return "Not palindrome"
        i+=1
        j-=1
    return "Palindrome"
tpl=tuple(map(int,input().split()))
palin(tpl)

                
```

```python
#Find the second largest element.
def largest(tpl):
    largest=tpl[0]
    second=tpl[0]
    for i in tpl:
        if i>largest:
            second=largest
            largest=i
        elif i>second and i!=largest:
            second=i
    return second 
tpl=tuple(map(int,input().split()))
largest(tpl)
```

```python
#Find the second smallest element.
def smallest(tpl):
    smallest=tpl[0]
    second=tpl[0]
    for i in tpl:
        if i<smallest:
            second=smallest
            smallest=i
        elif i<second and i!=smallest:
            second=i
    return second

tpl=tuple(map(int,input().split()))
smallest(tpl)  
```

```python
#Remove duplicate elements.
def dupli(tpl):
    new=[]
    for i in tpl:
        if i not in new:
            new.append(i)
    return tuple(new)
tpl=tuple(map(int,input().split()))
dupli(tpl)  
```

```python
#Find common elements between two tuples.
def common(t1,t2):
    new=[]
    for i in t1:
        for j in t2:
            if i==j:
                new.append(i)
    return tuple(new)
t1=tuple(map(int,input().split()))
t2=tuple(map(int,input().split()))
common(t1,t2)
```

```python
#Find elements present in the first tuple but not the second.
def first(t1,t2):
    new=[]
    for i in t1:
        found=False
        for j in t2:
            if i==j:
                found=True
                break
        if found==False:
            new.append(i)
    return tuple(new)
t1=tuple(map(int,input().split()))
t2=tuple(map(int,input().split()))
first(t1,t2)
```

```python
#Merge two tuples without duplicates.
def merge(t1,t2):
    new=[]
    for i in t1:
        if i not in new:
            new.append(i)
    for j in t2:
        if  j not in new:
            new.append(j)
    return tuple(new)
t1=tuple(map(int,input().split()))
t2=tuple(map(int,input().split()))
merge(t1,t2)
        
```

```python
#Find the frequency of every element.
def frequency(tpl):
    done=[]
    for i in tpl:
        if i not in done:
            count=0
            for j in tpl:
                if i==j:
                    count+=1
            print(i,count)
            done.append(i)
tpl=tuple(map(int,input().split()))
frequency(tpl)


        
```

```python
#Find the element with the highest frequency.
def freq(tpl):
    highest=0
    element=0
    for i in tpl:
        count=0
        for j in tpl:
            if i==j:
                count+=1
        if count>highest:
            highest=count
            element=i
    print(element,highest)
tpl=tuple(map(int,input().split()))
freq(tpl)   
    
```

```python
#Reverse a tuple without using [::-1].
def reverse(tpl):
    new=[]
    i=len(tpl)-1
    while i>=0:
        new.append(tpl[i])
        i-=1
    return tuple(new)
tpl=tuple(map(int,input().split()))
reverse(tpl) 
```

```python
#Check whether two tuples contain the same elements.
def same(t1,t2):
    if len(t1)!=len(t2):
        return "Not same"
    for i in t1:
        if i not in t2:
            return "Not same elements"
    return "Same Elements"
t1=tuple(map(int,input().split()))
t2=tuple(map(int,input().split()))
same(t1,t2)
```

```python
#Find the difference between the largest and smallest elements.
def diff(tpl):
    large=tpl[0]
    small=tpl[0]
    for i in tpl:
        if i>large:
            large=i
        if i<small:
            small=i
    difference=large-small
    return difference
tpl=tuple(map(int,input().split()))
diff(tpl)
```

```python
#Count how many elements are greater than a given number.
def greater(tpl,number):
    count=0
    for i in tpl:
        if i>number:
            count+=1
    return count
tpl=tuple(map(int,input().split()))
number=int(input("num:"))
greater(tpl,number)
```

```python
#Replace all occurrences of a given element with another element.
def occur(tpl,old,new):
    lst=[]
    for i in tpl:
        if i==old:
            lst.append(new)
        else:
            lst.append(i)
    return tuple(lst)
tpl=tuple(map(int,input().split()))
old=int(input("old:"))
new=int(input("new:"))
occur(tpl,old,new) 
```

```python
#Find the element that occurs only once.
def occur(tpl):
    for i in tpl:
        count=0
        for j in tpl:
            if i==j:
                count+=1
        if count==1:
            return i

tpl=tuple(map(int,input().split()))
occur(tpl)
```

```python
#Find whether all elements in a tuple are positive.
def pos(tpl):
    for i in tpl:
        if i<=0:
            return "Not all Positive"
    return "All positive"
tpl=tuple(map(int,input().split()))
pos(tpl)
```
