# Hierarchical Inheritance

```python
class animal:
    def eat(self):
        print("Eating")
class dog(animal):
    def bark(self):
        print("Barking")
class cat(animal):
    def meow(self):
        print("Meowing")
d=dog()
d.bark()
d.eat()
```

```python
c=cat()
c.meow()
c.eat()
```

```python
class person:
    def __init__(self,name):
        self.name=name
class student(person):
    def study(self):
        print(self.name)
        print("Studying")
class employee(person):
    def work(self):
        print(self.name)
        print("working")
s=student("Mythri")
s.study()
e=employee("Mythri")
e.work()
```

```python
class vehicle:
    def __init__(self,brand):
        self.brand=brand
class car(vehicle):
    def __init__(self,brand,doors):
        super().__init__(brand)
        self.doors=doors
    def display(self):
        print(self.brand)
        print(self.doors)
class bike(vehicle):
    def __init__(self,brand,engine):
        super().__init__(brand)
        self.engine=engine
    def display1(self):
        print(self.brand)
        print(self.engine)

c=car("Toyota",4)
c.display()
```

```python
b=bike("Honda",150)
b.display1()
```

```python
class employee:
    def __init__(self,name,salary):
        self.name=name
        self.salary=salary
class developer(employee):
    def bonuses(self):
        self.salary/=10
        print(self.salary)
class manager(employee):
    def manager_bonus(self):
        self.salary=self.salary*20/100
        print(self.salary)
d=developer("Mythri",60000)
d.bonuses()
```

```python
m=manager("Mythri",60000)
m.manager_bonus()
```

```python
class student:
    def __init__(self,name,marks):
        self.name=name
        self.marks=marks
class engineeringstudent(student):
    def passing(self):
        if self.marks>=75:
            print("Engineering: Distinction")
        else:
            print("Engineering: Regular")
class artsstudent(student):
    def arts(self):
        if self.marks>=60:
            print("Arts: Distinction")
        else:
            print("Arts: Regular")
s=engineeringstudent("Mythri",80)
s.passing()
```

```python
a=artsstudent("Mythri",80)
a.arts()
```
