```python
#Find the average of all elements.
#Method 2 using func
def average(lst,i):
    total=0
    i=0
    while i < len(lst):
        total+=lst[i]
        i+=1
    avg=total/len(lst)
    print(avg)
lst=list(map(int,input().split()))
average(lst,0)
```

```python
#Find the largest element without using max().
#Method 3: using recursion
def maximum(lst,i):
    if i==len(lst)-1:
        return lst[i]
    max_rest=maximum(lst,i+1)
    if lst[i]>max_rest:
        return lst[i]
    else:
        return max_rest
lst=list(map(int,input().split()))
maximum(lst,0)
```

```python
#Reverse every inner list without using reverse() or [::-1].
def reverse(lst):
    for i in range(len(lst)):
        left=0
        right=len(lst[i])-1
        while left<right:
            lst[i][left],lst[i][right]=lst[i][right],lst[i][left]
            left+=1
            right-=1
    return lst
lst= [[1, 2, 3], [4, 5], [6, 7, 8]]
reverse(lst)
```

```python
#Flatten a nested list into a single list.
def flatten(lst):
    result=[]
    for i in range(len(lst)):
        for j in range(len(lst[i])):
            result.append(lst[i][j])
    return result
lst=[[1, 2], [3, 4], [5, 6]]
flatten(lst)
```

```python
#Count the number of positive, negative and zero elements.
def count(lst):
    pos=0
    neg=0
    zero=0
    for i in range(len(lst)):
        for j in range(len(lst[i])):
            if lst[i][j]>0:
                pos+=1
            elif lst[i][j]==0:
                zero+=1
            else:
                neg+=1
    return pos,neg,zero
lst=[[2, -3, 0],
     [5, -1, 7],
     [0, -4, 8]]

count(lst)
```

```python
#Add two matrices
def addmatrix(m1,m2):
    result=[]
    for i in range(len(m1)):
        row=[]
        for j in range(len(m1[i])):
            row.append(m1[i][j]+m2[i][j])
        result.append(row)
    return result
m1= [[1, 2],
     [3, 4]]
m2= [[5, 6],
     [7, 8]]
addmatrix(m1,m2)
```

```python
#Subtract two matrices.
def submatrix(m1,m2):
    result=[]
    for i in range(len(m1)):
        row=[]
        for j in range(len(m1[i])):
            row.append(m1[i][j]-m2[i][j])
        result.append(row)
    return result
m1=[[5, 7],
    [9, 4]]
m2=[[1, 2],
    [3, 1]]
submatrix(m1,m2)
```

```python
#Find the sum of both diagonals.
def diagonals(lst):
    sum=0
    for i in range(len(lst)):
        for j in range(len(lst[i])):
            if i==j:
                sum+=lst[i][j]
            if i+j==len(lst)-1:
                sum+=lst[i][j]
            
    return sum
lst=[[1, 2, 3],
     [4, 5, 6],
     [7, 8, 9]]
diagonals(lst)
```

```python
#Find the difference between the two diagonal sums.
def diffsum(lst):
    sum1=0
    sum2=0
    for i in range(len(lst)):
        for j in range(len(lst[i])):
            if i==j:
                sum1+=lst[i][j]
            if i+j==len(lst)-1:
                sum2+=lst[i][j]
    difference=sum1-sum2
    return difference
lst=[[1, 2, 3],
     [4, 5, 6],
     [7, 8, 9]]
diffsum(lst)    
```

```python
#Check whether a matrix is a diagonal matrix.
def diagonal(lst):
    for i in range(len(lst)):
        for j in range(len(lst[i])):
            if i!=j:
                if lst[i][j]!=0:
                    return False
    return True
lst=[[5, 0, 0],
     [0, 3, 0],
     [0, 0, 7]]
diagonal(lst)
```

```python
#Check whether a matrix is an identity matrix.
def identity(lst):
    for i in range(len(lst)):
        for j in range(len(lst[i])):
            if i==j:
                if lst[i][j]!=1:
                    return False
            if i!=j:
                if lst[i][j]!=0:
                    return False
    return True
lst=[[1, 0, 0],
     [0, 1, 0],
     [0, 0, 1]]
identity(lst)
                    
```

```python
#Check whether a matrix is symmetric.
def symmetric(lst):
    for i in range(len(lst)):
        for j in range(len(lst[i])):
            if lst[i][j]!=lst[j][i]:
                return False
    return True
lst=[[1, 2, 3],
     [4, 5, 6],
     [7, 8, 9]]
symmetric(lst)
```

```python
#Find the transpose of a matrix.
def transpose(matrix):
    result=[]
    for j in range(len(matrix[0])):
        row=[]
        for i in range(len(matrix)):
            row.append(matrix[i][j])
        result.append(row)
    return result
matrix=[[1, 2, 3],
        [4, 5, 6]]
transpose(matrix)
```

```python
#Print all boundary elements of a matrix.
def boundary(lst):
    result=[]
    for i in range(len(lst)):
        for j in range(len(lst[i])):
            if i==0 or i==len(lst)-1 or j==0 or j==len(lst[i])-1:
                result.append(lst[i][j])
    return result
lst = [[1, 2, 3],
       [4, 5, 6],
       [7, 8, 9]]
boundary(lst)
```
