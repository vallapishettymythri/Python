```python
#Q1.Write a  Python program to check if each number is prime in a given list of
#numbers. Return True if all numbers are prime otherwise False.
#using func
def primeno(lst):
    prime=True
    for i in range(len(lst)):
        if lst[i]<2:
            prime=False
            break
        for j in range(2,lst[i]):
            if lst[i]%j==0:
                prime=False
                break
    if prime:
        return True
    else:
        return False
lst=list(map(int,input().split()))
primeno(lst)
            
```

```python
#using recursion
def primeno(lst,i,j):
    if i==len(lst):
        return True
    if lst[i]<2:
        return False
    if j==lst[i]:
        return primeno(lst,i+1,2)
    if lst[i]%j==0:
        return False
    return primeno(lst,i,j+1)
lst=list(map(int,input().split()))
primeno(lst,0,2)
```

```python
#Q2. Write a  Python program to generate all permutations of a list in Python.
def permutations(lst,result=[]):
    if len(lst)==0:
        print(result)
        return
    for i in range(len(lst)):
        current=lst[i]
        remaining=lst[:i]+lst[i+1:]
        permutations(remaining,result+[current])
lst=[1,2,3]
permutations(lst)
```

```python
#3.Write a Python program that uses the Sieve of Eratosthenes method to
#compute prime numbers up to a specified number.
def sieve(n):
    prime=[True]*(n+1)
    prime[0]=False
    prime[1]=False
    for i in range(2,n+1):
        if prime[i]:
            for j in range(i*2, n+1,i):
                prime[j]=False
    result=[]

    for i in range(2,n+1):
        if prime[i]:
            result.append(i)
    return result
n=int(input("n:"))
sieve(n)
```

```python
#Write a Python program to compute the product of the odd digits in a
#given number, or 0 if there aren't any.
def odd_digi(n):
    odd=1
    while n>0:
        digi=n%10
        if digi%2!=0:
            odd*=digi
        else:
            return 0
        n//=10
    return odd
n=int(input("n:"))
odd_digi(n)
```

```python
#Q6. Write a Python program to find the largest integer divisor of a number n
#that is less than n.
def divisor(n):
    for i in range(n-1,0,-1):
        if n%i==0:
            return i
n=int(input("n:"))
divisor(n)
```

```python
#Q7. Write a Python program to sort the numbers in a given list by the sum of
#their digits.
def sum_digi(n):
    total=0
    while n>0:
        digi=n%10
        total+=digi
        n//=10

    return total
def sort(lst):
    for i in range(len(lst)):
        for j in range(i+1,len(lst)):
            if sum_digi(lst[i])>sum_digi(lst[j]):
                lst[i],lst[j]=lst[j],lst[i]
    return lst
lst = list(map(int, input().split()))
sort(lst)
        
```

```python
#Write a Python program to determine which triples sum to zero from a
#given list of lists.
def triples(lst):
    result=[]
    for i in range(len(lst)):
        if lst[i][0]+lst[i][1]+lst[i][2]==0:
            result.append(True)
        else:
            result.append(False)
    return result

lst = [[1343532, -2920635, 332],
       [-27, 18, 9],
       [4, 0, -4],
       [2, 2, 2],
       [-20, 16, 4]]
triples(lst)
    
```

```python
#Q9. Write a Python program to sort numbers based on strings.
def sort_word(lst):
    numbers={
        "zero": 0,
        "one": 1,
        "two": 2,
        "three": 3,
        "four": 4,
        "five": 5,
        "six": 6,
        "seven": 7,
        "eight": 8,
        "nine": 9
    }

    for i in range(len(lst)):
        for j in range(i+1,len(lst)):
            if numbers[lst[i]]>numbers[lst[j]]:
                lst[i],lst[j]=lst[j],lst[i]
    return lst
lst=input().split()
sort_word(lst)
```

```python
#Write a Python program to find all even palindromes up to n.
def palindrome(n):
    original=n
    reverse=0

    while n>0:
        digit=n%10
        reverse=reverse*10+digit
        n//=10
    return original==reverse

def even_palindrome(n):
    result=[]
    for i in range(n+1):
        if i%2==0:
            if i==0 or palindrome(i):
                result.append(i)
    return result
n=int(input("n:"))
even_palindrome(n)
```

```python
#Write a Python program to find a list of all numbers that are adjacent to a
#prime number in the list, sorted without duplicates.
def prime(n):
    if n<2:
        return False
    for i in range(2,n):
        if n%i==0:
            return False
    return True
def adjacent(lst):
    result=[]
    for  i in range(len(lst)):
        if prime(lst[i]):
            if i>0:
                result.append(lst[i-1])
            if i< len(lst)-1:
                result.append(lst[i+1])
    new=[]
    for x in result:
        if x not in new:
            new.append(x)
    for i in range(len(new)):
        for j in range(i + 1, len(new)):
            if new[i] > new[j]:
                new[i], new[j] = new[j], new[i]  
    return new
lst = list(map(int, input().split()))
adjacent(lst)
            
```

```python
#gnome sort
def gnome(lst):
    i=0
    while i<len(lst):
        if i==0 or lst[i]>=lst[i-1]:
            i+=1
        else:
            lst[i],lst[i-1]=lst[i-1],lst[i]
            i-=1
    return lst
lst = list(map(int, input().split()))
gnome(lst)
```
