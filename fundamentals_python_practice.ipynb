```python
def balanced(n):
    even=0
    odd=0
    while n>0:
        digit=n%10
        if digit%2==0:
            even+=digit
        else:
            odd+=digit
        n//=10
    if even==odd:
        return True
    else:
        return False
n=int(input())
balanced(n)
```

```python
def character_occur(s):
    for i in range (len(s)):
        count=0
        ch=""
        for j in range(len(s)):
            if s[i]==s[j]:
                count+=1
                ch+=s[i]
        if count==1:
            return ch
s=input()
character_occur(s)  
```

```python
def neighbours(n,a):
    current=1
    longest=1
    for i in range(1,n):
        if a[i]>a[i-1]:
            current+=1

            if current > longest:
                longest=current
        else:
            current=1
    return longest
n = int(input())
a = list(map(int, input().split()))
neighbours(n,a)
```

```python
def move(lst):
    new=[]
    zero_count=0
    for i in lst:
        if i!=0:
            new.append(i)
        else:
            zero_count+=1
    for i in range(zero_count):
        new.append(0)
    return new
lst=[0,1,0,3,12,0,5,7]
move(lst)
```

```python
def occurs(s):
    high=0
    ch=""
    words=s.split()
    for i in words:
        count=0
        for j in words:
            if i==j:
                count+=1

                if count> high:
                    high=count
                    ch=i
    return ch,high
s="hello world hello python world hello"
occurs(s)
```

```python
def no_dup(lst):
    new=[]
    for i in lst:
        if i not in new:
            new.append(i)
    return new
lst=list(map(int,input().split()))
no_dup(lst)
```

```python
def anagarams(a,b):
    if len(a)!=len(b):
        return "Not anagrams"
    counta={}
    countb={}
    for ch in a:
        if ch in counta:
            counta[ch]+=1
        else:
            counta[ch]=1
    for ch in b:
        if ch in countb:
            countb[ch] += 1
        else:
            countb[ch] = 1
    if counta==countb:
        return "Anagrams"
    else:
        return "Not anagrams"
a="listen"
b="silent"
anagarams(a,b)       
        
```

```python
def consecutive(lst):
    count = 1
    longest = 1

    for i in range(1, len(lst)):
        if lst[i] == lst[i - 1]:
            count += 1
        else:
            count = 1

        if count > longest:
            longest = count

    return longest


lst = [5, 5, 7, 7, 7, 2, 2, 9]

print(consecutive(lst))
```

```python
def subsequence(s,sub):
    j=0

    for i in s:
        if j<len(sub) and i==sub[j]:
            j+=1
    if j==len(sub):
        return "Subsequence"
    else:
        return "Not subsequence"

s = input()
sub = input()
subsequence(s,sub)
```

```python
def employee(lst):
    high=0
    emp=""
    for i in range(len(lst)):
        count=0
        for j in range(len(lst)):
            if lst[i]==lst[j]:
                count+=1
        if count>high:
            high=count
            emp+=lst[i]
    return emp,count

lst=["A", "B", "A", "C", "B", "A"]
employee(lst)
```

```python
def product_purchase(lst):
    dict={}
    for product,quantity in lst:
        if product not in dict:
            dict[product]=quantity
        elif product in dict:
            dict[product]+=quantity
    return dict
lst=[
    ("laptop", 2),
    ("mouse", 3),
    ("laptop", 1),
    ("keyboard", 2)
]

product_purchase(lst)
```

```python
dict={
    "CSE": ["A", "B", "C"],
    "ECE": ["D", "E"],
    "AIML": ["F", "G"]
}
student="E"
def contain_student(dict,student):
    for branch,name in dict.items():
        for i in name:
            if i==student:
                return branch
                
contain_student(dict,student)
```

```python
import re
f1=open("file.txt","w")
f1.write("Hello everyone\n")
f1.write("Contact me at mythri@gmail.com\n")
f1.write("This is a normal line\n")
f1.write("Email: rahul123@yahoo.com\n")
f1.write("Thank you\n")
f1.close()


f1=open("file.txt","r")
data=f1.read()
f1.close()

def email(data):
    pattern="[a-zA-Z0-9]+\@[a-z]+\.[a-z]+"
    count=0

    lines=data.splitlines()
    for i in lines:
        if re.search(pattern,i):
            count+=1
    return count

email(data)  
    
```
