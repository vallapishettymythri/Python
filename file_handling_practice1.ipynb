```python
#Q2. You are asked to design a file system that allows you to create new paths
#and associate them with different values.
#The format of a path is one or more concatenated strings 
class filesystem:
    def __init__(self):
        self.path={}
    def createpath(self,path,value):
        if path in self.path:
            return False
        parent=path.rsplit("/",1)[0] #split from right side only once

        if parent!="" and parent not in self.path:
            return False
        self.path[path]=value
        return True
    def get(self,path):
        if path in self.path:
            return self.path[path]
        return -1
fs = filesystem()

print(fs.createpath("/leetcode", 1))
print(fs.createpath("/leetcode/problems", 2))
print(fs.get("/leetcode"))
print(fs.get("/leetcode/problems"))
print(fs.get("/unknown"))
```

```python
#Find and Replace Text
#Write a script that finds a specific word in a file and replaces it with another
#word.
def findreplace(data,find,replace):
    data=data.replace(find,replace)
    return data
    
f1=open("file.txt","w")
f1.write("I like Python.\n")
f1.write("Python is easy to learn.\n")
f1.write("I practice Python every day.")
f1.close()

f1=open("file.txt","r")
data=f1.read()
f1.close()

find=input("word to find:")
replace=input("word to replace:")
print(findreplace(data,find,replace))
```

```python
#Remove Blank Lines
#Create a script to remove all blank lines from a file.
def blank(data):
    s=""
    lines=data.splitlines()
    for i in lines:
        if i!="":
            s+=i
            s+="\n"
    return s
            

f1 = open("file.txt", "w")

f1.write("Hello Python.\n")
f1.write("\n")
f1.write("This is a file.\n")
f1.write("\n")
f1.write("I am learning file handling.\n")
f1.write("\n")
f1.write("Thank you.")

f1.close()

f1=open("file.txt")
data=f1.read()
f1.close()

print(blank(data))
```

```python
#.Read a File in Reverse Order
#Print the lines of a file in reverse order (last line first).
def reverse(data):
    s=""
    lines=data.splitlines()
    for i in range(len(lines)-1,-1,-1):
        if lines[i]!="":
            s+=lines[i]+"\n"
    return s


f1 = open("file.txt", "w")

f1.write("Hello Python.\n")
f1.write("\n")
f1.write("This is a file.\n")
f1.write("\n")
f1.write("I am learning file handling.\n")
f1.write("\n")
f1.write("Thank you.")

f1.close()

f1=open("file.txt")
data=f1.read()
f1.close()

print(reverse(data))
```

```python
#Q10.Line Frequency Counter
#Count the frequency of each line in a file (useful for analyzing logs or data
#files).
def freq(data):
    dict={}
    lines=data.splitlines()
    for i in lines:
        if i not in dict:
            dict[i]=1
        elif i in dict:
            dict[i]+=1
    return dict


f1 = open("file.txt", "w")

f1.write("Python\n")
f1.write("Java\n")
f1.write("Python\n")
f1.write("C++\n")
f1.write("Java\n")
f1.write("Python\n")

f1.close()


f1=open("file.txt")
data=f1.read()
f1.close()

freq(data)
```
