```python
#Add Member(s) to a Set
#Write a Python program to add member(s) to a set.
def addmem(s,members):
    for i in members:
        s.add(i)
    return s
s={10,20,30}
members={40,50}
addmem(s,members)
```

```python
#Remove Item(s) from a Given Set
#Write a Python program to remove item(s) from a given set.
def removeitem(s,item):
    for i in item:
        if i in s:
            s.remove(i)
    return s
s = {10, 20, 30, 40, 50}
item = {20, 40}
removeitem(s,item)
```

```python
#Remove an Item from a Set if Present
#Write a Python program to remove an item from a set if it is present in the set.
def my_remove(s,item):
    ss=set()
    for i in s:
        if i!=item:
            ss.add(i)
    return ss
s = {10, 20, 30, 40}
item = 30
my_remove(s,item)
```

```python
#Create an Intersection of Sets
#Write a Python program to create an intersection of sets.
def intersection(A,B):
    s=set()
    for i in A:
        if i in B:
            s.add(i)
    return s
A = {10, 20, 30, 40}
B = {30, 40, 50, 60}
intersection(A,B)
```

```python
#Create a Union of Sets
#Write a Python program to create a union of sets.
def unions(A,B):
    s=set()
    for i in A:
        s.add(i)
    for i in B:
        s.add(i)
    return s
A = {10, 20, 30}
B = {30, 40, 50}
unions(A,B)
```

```python
#Create Set Difference
#Write a Python program to create set difference.
def difference(A,B):
    s=set()
    for i in A:
        if i not in B:
            s.add(i)
    return s
A = {10, 20, 30, 40}
B = {30, 40, 50, 60}
difference(A,B)
```

```python
#Create a Symmetric Difference
#Write a Python program to create a symmetric difference.
def symmetric(A,B):
    s=set()
    for i in A:
        if i not in B:
            s.add(i)
    for i in B:
        if i not in A:
            s.add(i)
    return s
A = {10, 20, 30, 40}
B = {30, 40, 50, 60}
symmetric(A,B)
```

```python
#Check if a Set is a Subset of Another Set
#Write a Python program to check if a set is a subset of another set.
def subset(A,B):
    for i in A:
        if i not in B:
            return False
    return True
A = {10, 20}
B = {10, 20, 30, 40}
subset(A,B)
```

```python
#Find Maximum and Minimum Values in a Set
#Write a Python program to find the maximum and minimum values in a set.
def maxmini(s):
    s=list(s)
    maxi=mini=s[0]
    for i in s:
        if i>maxi:
            maxi=i
        if i<mini:
            mini=i
    return maxi,mini
s = {10, 25, 5, 40, 15}
maxmini(s)      
```

```python
#Check if a Given Value is Present in a Set
#Write a Python program to check if a given value is present in a set or not.
def check(s,val):
    for i in s:
        if val==i:
            return "Present"
    return "Not present"
s = {10, 20, 30, 40, 50}
val = 30
check(s,val)
```

```python
#Check if Two Sets Have No Elements in Common
#Write a Python program to check if two given sets have no elements in common.
def common(A,B):
    for i in A:
        if i in B:
             return "Common elements present"
    return "No common elements"
A = {10, 20, 30}
B = {40, 50, 60}
common(A,B)
```

```python
#Check if a Set is a Superset
#Write a Python program to check if a given set is a superset of itself and a
#superset of another given set.
def superset(A,B):
    for i in B:
        if i not in A:
            return "Not Superset"
    return "Superset"
A = {10, 20, 30, 40}
B = {20, 30}
superset(A,B)
```

```python
#Find Elements in a Set Not in Another
#Write a Python program to find elements in a given set that are not in another set.
def elements(A,B):
    s=set()
    for i in A:
        if i not in B:
            s.add(i)
    return s
A = {10, 20, 30, 40, 50}
B = {30, 40, 60}
elements(A,B)
```
