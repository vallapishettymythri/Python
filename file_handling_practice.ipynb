```python
#Remove Empty Lines
def remove(data):
    lines=data.splitlines()
    res=[]
    for i in lines:
        if i!="":
            res.append(i)
    return res

f1=open("file.txt","w")
f1.write("Python\n")
f1.write("\n")
f1.write("Java\n")
f1.write("\n")
f1.write("C++\n")
f1.write("\n")
f1.write("JavaScript\n")
f1.close()

f1=open("file.txt","r")
data=f1.read()
f1.close()

remove(data)
```

```python
#Convert File Content to Uppercase- method one using upper
def uppercase(data):
    lines=data.splitlines()
    res=[]
    for i in lines:
        res.append(i.upper())
    return res
            

f1=open("file.txt","w")
f1.write("Python is easy\n")
f1.write("Java is powerful")
f1.close()

f1=open("file.txt")
data=f1.read()
f1.close()

uppercase(data)
```

```python
#Convert File Content to Uppercase- method 2 usinf chr,ord
def uppercase(data):
    lines=data.splitlines()
    res=[]
    for i in lines:
        new_line=""
        for j in i:
            if "a"<= j <="z":
                new_line+=chr(ord(j)-32)
            else:
                new_line+=j
        res.append(new_line)
    return res
            

f1=open("file.txt","w")
f1.write("Python is easy\n")
f1.write("Java is powerful")
f1.close()

f1=open("file.txt")
data=f1.read()
f1.close()

uppercase(data)
```

```python
#Take a word from the user and replace every occurrence of it with another word.
def replaceword(word,data,new):
    lines=data.splitlines()
    res=[]
    for i in lines:
        if word in i:
            i=i.replace(word,new)
        res.append(i)
    return res
    
            

f1=open("file.txt","w")
f1.write("I like Python\n")
f1.write("Python is easy\n")
f1.write("I am learning Python")
f1.close()

f1=open("file.txt")
data=f1.read()
f1.close()

word="Python"
new="Java"
replaceword(word,data,new)
```

```python
#Count Uppercase and Lowercase Characters
def uplw(data):
    words=data.split()
    upper=0
    lower=0
    for i in words:
        for j in i:
            if j.isupper():
                upper += 1
            elif j.islower():
                lower += 1
    return upper,lower
                

f1=open("file.txt","w")
f1.write("python JAVA")
f1.close()

f1=open("file.txt")
data=f1.read()
f1.close()

uplw(data)
```

```python
#Compare Two Files
def compare(a,b):
    if a==b:
        return True
    return False
            

f1=open("file.txt","w")
f1.write("Python\n")
f1.write("Java\n")
f1.write("C++")
f1.close()

f1=open("file.txt")
a=f1.read()
f1.close()

f2=open("data.txt","w")
f2.write("Python\n")
f2.write("Java\n")
f2.write("C++")
f2.close()

f2=open("data.txt")
b=f2.read()
f2.close()

compare(a,b)
```

```python
#Find Differences Between Two Files
def difference(a,b):
    linea=a.splitlines()
    lineb=b.splitlines()
    res=[]
    for i in linea:
        if i not in lineb:
            res.append(i)
    for i in lineb:
        if i not in linea:
            res.append(i)
    return res
            

f1=open("file.txt","w")
f1.write("Python\n")
f1.write("Java\n")
f1.write("C++")
f1.close()

f1=open("file.txt")
a=f1.read()
f1.close()

f2=open("data.txt","w")
f2.write("Python\n")
f2.write("Java\n")
f2.write("Ruby")
f2.close()

f2=open("data.txt")
b=f2.read()
f2.close()

difference(a,b)
```

```python
# Find Common Lines
def commonline(a,b):
    linea=a.splitlines()
    lineb=b.splitlines()
    res=[]
    for i in range(len(linea)):
        for j in range(len(lineb)):
            if linea[i]==lineb[j]:
                res.append(linea[i])
    return res

f1=open("file.txt","w")
f1.write("Python\n")
f1.write("Java\n")
f1.write("C++")
f1.close()

f1=open("file.txt")
a=f1.read()
f1.close()

f2=open("data.txt","w")
f2.write("Python\n")
f2.write("Java\n")
f2.write("Ruby")
f2.close()

f2=open("data.txt")
b=f2.read()
f2.close()

commonline(a,b)
```

```python
# Group Students by Grade
def grade(data):
    lines=data.splitlines()
    dict={}
    for i in lines:
        name,marks=i.split(",")
        marks=int(marks)
        if marks>90:
            grade="A"
        elif 75<marks<89:
            grade="B"
        elif 60<marks<74:
            grade="C"
        elif marks<60:
            grade="D"
        dict[name]=grade
    return dict

f1=open("file.txt","w")
f1.write("Rahul,35\n")
f1.write("Priya,92\n")
f1.write("Arjun,46\n")
f1.write("Sneha,88")
f1.close()

f1=open("file.txt")
data=f1.read()
f1.close()

grade(data)
```

```python
#Extract Email Addresses
import re
def email(data):
    pattern="[a-zA-z0-9]+\@[a-z]+\.[a-z]+"
    return re.findall(pattern,data)
    

f1=open("file.txt","w")
f1.write("Contact john@gmail.com for details.\n")
f1.write("Send your resume to hr@company.com.\n")
f1.write("Invalid email: hello@")
f1.close()

f1=open("file.txt")
data=f1.read()
f1.close()

email(data)
```

```python
#Extract Indian Mobile Numbers
def indiamobile(data):
    pattern="\+91[6-9]\d{9}"
    return re.findall(pattern,data)

f1=open("file.txt","w")
f1.write("Call +918790289211\n")
f1.write("Office: +919876543210\n")
f1.write("Invalid: +911234567890")
f1.close()

f1=open("file.txt")
data=f1.read()
f1.close()

indiamobile(data)
```

```python
#Extract URLs
def urls(data):
    pattern="https\:\/\/[a-z]+\.[a-z]+"
    return re.findall(pattern,data)

f1=open("file.txt","w")
f1.write("Visit https://google.com\n")
f1.write("Visit https://github.com\n")
f1.write("Invalid: hello.com")
f1.close()

f1=open("file.txt")
data=f1.read()
f1.close()

urls(data)
```

```python
#Remove Special Characters
def special(data):
    pattern="[!@#$%^&*]"
    return re.sub(pattern,"",data)

f1=open("file.txt","w")
f1.write("Hello@Python\n!")
f1.write("Python#Programming\n")
f1.write("C++ is great!!!\n")
f1.close()

f1=open("file.txt")
data=f1.read()
f1.close()

print(special(data))
```

```python
#Extract Words Starting With Capital Letters
def capital(data):
    pattern="[A-Z][a-z]+"
    return re.findall(pattern,data)

f1=open("file.txt","w")
f1.write("Python is Easy\n")
f1.write("Java is Powerful\n")
f1.write("I Love Programming")
f1.close()

f1=open("file.txt")
data=f1.read()
f1.close()

capital(data)
```

```python
#Replace Multiple Spaces With One Space
def space(data):
    pattern=" +"
    return re.sub(pattern," ",data)

f1=open("file.txt","w")
f1.write("Python  is  Easy\n")
f1.write("Java  is  Powerful\n")
f1.close()

f1=open("file.txt")
data=f1.read()
f1.close()

print(space(data))
```

```python
#Remove Comments From a Python File
def comments(data):
    pattern="#.*"
    return re.sub(pattern,"",data)

f1=open("program.py","w")
f1.write("x = 10\n")
f1.write("# this is a comment\n")
f1.write("y = 20\n")
f1.write("print(x + y) # calculate sum")
f1.close()

f1=open("program.py")
data=f1.read()
f1.close()

print(comments(data))

```

```python
#Extract Quoted Words
def qouted(data):
    pattern='"[A-Za-z]+"'
    return re.findall(pattern,data)

f1=open("file.txt","w")
f1.write('I said "hello"\n')
f1.write('She said "Python"\n')
f1.write('He said "Java"\n')
f1.close()

f1=open("file.txt")
data=f1.read()
f1.close()

qouted(data)
```

```python
#Log File Error Extraction
def error(data):
    pattern="ERROR.*"
    match=re.findall(pattern,data)
    f2=open("data.txt","w")
    for i in match:
        f2.write(i+"\n")
    f2.close()

    f2=open("data.txt")
    data1=f2.read()
    f2.close()
    print(data1)

f1=open("file.txt","w")
f1.write("INFO Server started\n")
f1.write("ERROR Database failed\n")
f1.write("INFO User logged in\n")
f1.write("ERROR File not found\n")
f1.write("WARNING Low memory")
f1.close()

f1=open("file.txt")
data=f1.read()
f1.close()

error(data)
```

```python
#Analyze a Log File
def analyze(data):
    pattern=r"[A-Z]+\b"
    match=re.findall(pattern,data)
    dict={}
    for i in match:
        if i not in dict:
            dict[i]=1
        else:
            dict[i]+=1
    return dict
        

f1=open("file.txt","w")
f1.write("INFO Server started\n")
f1.write("ERROR Database failed\n")
f1.write("INFO User logged in\n")
f1.write("ERROR File not found\n")
f1.write("WARNING Low memory")
f1.close()

f1=open("file.txt")
data=f1.read()
f1.close()
analyze(data)
```

```python
#Password Validation From File
def password(data):
    pattern="[A-Za-z0-9@]{8,}"
    return re.findall(pattern,data)

    
f1=open("file.txt","w")
f1.write("Hello123\n")
f1.write("python\n")
f1.write("Python@123\n")
f1.write("abc123\n")
f1.write("Test@456")
f1.close()

f1=open("file.txt")
data=f1.read()
f1.close()

password(data)
```

```python
#Create a program using functions that generates report.txt. Mini file analyzer
def mini(data):
    dict={}
    lines=data.splitlines()
    linecount=0
    character=0
    for i in lines:
        linecount+=1
        for j in i:
            character+=1
    dict["Lines"]=linecount
    dict["Character"]=character
    words=data.split()
    wordcount=0
    for i in words:
        wordcount+=1
    dict["words"]=wordcount

    vowel="[aeiouAEIOU]"
    match=re.findall(vowel,data)
    vowelcount=0
    for i in match:
        vowelcount+=1
    dict["vowel"]=vowelcount

    digits="\d"
    digi=re.findall(digits,data)
    digicount=0
    for i in digi:
        digicount+=1
    dict["digits"]=digicount
    return dict
f1=open("file.txt","w")
f1.write("Python is easy to learn.\n")
f1.write("Python is powerful.\n")
f1.write("I love programming in Python.\n")
f1.write("Contact: test@gmail.com\n")
f1.write("Call: +918790289211")
f1.close()

f1=open("file.txt")
data=f1.read()
f1.close()

mini(data)
```
