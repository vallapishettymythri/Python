```python
#Find the sum of digits of a number.
n=int(input("n:"))
sum=0
count=0
temp=n
while temp>0:
    count+=1
    temp//=10
for i in range(count):
    digit=n%10
    sum+=digit
    n//=10
print(sum)
```

```python
#Find the largest digit in a number.
n=int(input("n:"))
large=n%10
count=0
temp=n
while temp>0:
    count+=1
    temp//=10
for i in range(count):
    digit=n%10
    if digit > large:
        large=digit
    n//=10
print(large)
```

```python
#Reverse a number using a for loop.
n = int(input("n:"))

temp = n
count = 0

while temp > 0:
    count += 1
    temp //= 10

reverse = 0

for i in range(count):
    digit = n % 10
    reverse = reverse * 10 + digit
    n //= 10

print(reverse)
```

```python
#Check whether a number is palindrome.
n = int(input("n:"))
original=n

temp = n
count = 0

while temp > 0:
    count += 1
    temp //= 10

reverse = 0

for i in range(count):
    digit = n % 10
    reverse = reverse * 10 + digit
    n //= 10
if original==reverse:
    print("is palindrome")
else:
    print("not palindrome")

```

```python
#Check whether a number is Armstrong.
n = int(input("n:"))
original=n
temp = n
count = 0
sum=0

while temp > 0:
    count += 1
    temp //= 10

for i in range(count):
    digit=n%10
    sum+=digit**count
    n//=10
print(sum)

if original==sum:
    print("Is amstrong")
else:
    print("Not amstrong")
```

```python
#Check whether a number is a Strong number.
n = int(input("n:"))
original=n
temp = n
count = 0
sum=0

while temp > 0:
    count += 1
    temp //= 10

for i in range(count):
    digit=n%10
    fact=1
    for j in range(1,digit+1):
        fact*=j
    sum+=fact
    n//=10
   
print(sum)
if original==sum:
    print("Is strong number")
else:
    print("Not strong number")
```

```python
#Print each digit of a number from left to right.
n=int(input("n:"))
count=0
temp=n
while temp>0:
    count+=1
    temp//=10
tens=10**(count-1)
for i in range(count):
    digit=n//tens
    print(digit)
    n%=tens
    tens//=10
```

```python
#Print each digit of a number from right to left.
n=int(input("n:"))
count=0
temp=n
while temp>0:
    count+=1
    temp//=10
for i in range(count):
    digit=n%10
    print(digit)
    n//=10
```

```python
#Count how many times a particular digit occurs.
n=int(input("n:"))
digit=int(input("digit:"))
count=0
occur=0
temp=n
while temp>0:
    count+=1
    temp//=10
for i in range(count):
    digits=n%10
    if digits==digit:
        occur+=1
    n//=10
print(occur)
```

```python
#Check whether all digits of a number are the same.
n=int(input("n:"))
count=0
temp=n
num=n%10
flag=True
while temp>0:
    count+=1
    temp//=10
for i in range(count):
    digit=n%10
    if digit!=num:
        flag=False
        break
    n//=10
if flag:
    print("All the digits are equal")
else:
    print("All the digits are not equal")

              
    
```

```python
#Find the second largest digit.
n=int(input("n:"))
count=0
temp=n
first=-1
second=-1
while temp>0:
    count+=1
    temp//=10
for i in range(count):
    digit=n%10
    if digit > first:
        second=first
        first=digit
    elif digit > second and digit!=first:
        second=digit
    n//=10
print(first,second)
```

```python
#Find the second smallest digit.
n=int(input("n:"))
count=0
temp=n
first=10
second=10
while temp>0:
    count+=1
    temp//=10
for i in range(count):
    digit=n%10
    if digit < first:
        second=first
        first=digit
    elif digit < second and digit!=first:
        second=digit
    n//=10
print(first,second)
```

```python
#Swap the first and last digit of a number.
n=int(input("n:"))
count=0
temp=n
while temp>0:
    count+=1
    temp//=10
last=n%10
tren=10**(count-1)
first=n//tren
middle=(n%tren)//10
print(last*tren+middle*10+first)
```

```python
#Find the middle digit of an odd-digit number.
n=int(input("n:"))
count=0
temp=n
while temp>0:
    count+=1
    temp//=10
middle=count//2
for i in range(middle):
    n//=10
    middle=-1
print(n%10)
```
