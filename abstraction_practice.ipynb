# Abstraction

```python
from abc import ABC,abstractmethod
```

```python
class animal(ABC):
    @abstractmethod
    def sound(self):
        pass
class dog(animal):
    def sound(self):
        print("Dog makes sound")
a=dog()
a.sound()
```

```python
class vehicle(ABC):
    @abstractmethod
    def start(self):
        pass
class car(vehicle):
    def start(self):
        print("Car starts with key")
class bike(vehicle):
    def start(self):
        print("Bike starts with self start")
```

```python
c=car()
c.start()
```

```python
b=bike()
b.start()
```

```python
class employee(ABC):
    def __init__(self,name):
        self.name=name
    @abstractmethod
    def work(self):
        pass
class developer(employee):
    def work(self):
        print(self.name,"Writes codes")
d=developer("Mythri")
d.work()
    
```

```python
class shape(ABC):
    def __init__(self,length,breadth):
        self.length=length
        self.breadth=breadth
    @abstractmethod
    def area(self):
        pass
class rectangle(shape):
    def area(self):
        print(self.length*self.breadth)
r=rectangle(10,5)
r.area()
```

```python
class bank(ABC):
    def __init__(self,amount):
        self.amount=amount
    @abstractmethod
    def interest(self):
        pass
class SBI(bank):
    def interest(self):
        Interest=self.amount*5/100
        print(Interest)
b=SBI(10000)
b.interest()
```
