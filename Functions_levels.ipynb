```python
#Function to check whether a number is a palindrome.
def palin(n):
    reverse=0
    temp=n
    while temp>0:
        digit=temp%10
        reverse = reverse * 10 + digit
        temp//=10
    if n==reverse:
        return "Is palindrome"
    else:
        return "Not palindrome"
n=int(input("n:"))
palin(n)
```

```python
#Function to check whether a number is an Armstrong number.
def amstrong(n):
    sum=0
    temp=n
    count=0
    original=n
    while temp>0:
        count+=1
        temp//=10
    for i in range(count):
        digit=n%10
        sum+=digit**count
        n//=10
    print(sum)
    if original==sum:
        print("Is armstrong")
    else:
        print("Not armstrong")
n=int(input("n:"))
amstrong(n)
    
```

```python
#Write factorial() and use it inside strong_number().
def fact(n):
    i=1
    fact=1
    while i<=n:
        fact*=i
        i+=1
    return fact
def strong_number(n):
    sum=0
    temp=n
    while temp>0:
        digit=temp%10
        sum+=fact(digit)
        temp//=10
    if sum==n:
        return "Strong number"
    else:
        return "Not strong number"
n=int(input("n:"))
strong_number(n)
    
```
