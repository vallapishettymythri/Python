### LEVEL 2 — Set Operations

```python
#Union of Two Sets
def unionset(A,B):
    s=set()
    for i in A:
        if i not in s:
            s.add(i)
    for i in B:
        if i not in s:
            s.add(i)
    return s
A = {1, 2, 3}
B = {3, 4, 5}
unionset(A,B)
```

```python
# Intersection
def inter(A,B):
    s=set()
    for i in A:
        for j in B:
            if i==j:
                s.add(i)
    return s
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
inter(A,B)
```

```python
# Difference A − B (A elements not present in B)
def difference(A,B):
    s=set()
    for i in A:
        if i not in B:
            s.add(i)
    return s
A = {1, 2, 3, 4}
B = {3, 4, 5}
difference(A,B)
```

```python
#Difference B − A
def difference(A,B):
    s=set()
    for i in B:
        if i not in A:
            s.add(i)
    return s
A = {1, 2, 3, 4}
B = {3, 4, 5}
difference(A,B)
```

```python
#Symmetric Difference(not in both)
def symmetric(A,B):
    s=set()
    for i in A:
        if i not in B:
            s.add(i)
    for i in B:
        if i not in A:
            s.add(i)
    return s

A = {1, 2, 3}
B = {3, 4, 5}
symmetric(A,B)
```

### LEVEL 3 — Set Problem Solving

```python
#Remove Duplicates from a List
def dupli(lst):
    s=set()
    for i in lst:
        s.add(i)
    return list(s)
lst=[10, 20, 10, 30, 20, 40]
dupli(lst)
```

```python
# Find Duplicate Elements
def finddupli(lst):
    s=set()
    for i in range (len(lst)):
        for j in range(i+1,len(lst)):
            if lst[i]==lst[j]:
                s.add(lst[i])
    return s
lst=[1, 2, 3, 2, 4, 1, 5]
finddupli(lst)
```

```python
#Find Elements Occurring Only Once
def  occur(lst):
    s=set()
    for i in range (len(lst)):
        count=0
        for j in range(len(lst)):
            if lst[i]==lst[j]:
                count+=1
        if count==1:
            s.add(lst[i])
    return s
lst=[1, 2, 2, 3, 4, 4, 5]
occur(lst)
```

```python
# Find Common Digits in Two Numbers
#using while
def commondigi(a,b):
    s1=set()
    s2=set()
    while a>0:
        digit=a%10
        s1.add(digit)
        a//=10
    while b>0:
        digit=b%10
        if digit in s1:
            s2.add(digit)
        b//=10
    return s2

a=123456
b=456789
commondigi(a,b)
```

### LEVEL 4 — Complex Set Problems

```python
#Find Elements Occurring in Exactly One List
def occur(A,B,C):
    s=set()
    for i in A:
        if i not in B and i not in C:
            s.add(i)
    for i in B:
        if i not in A and i not in C:
            s.add(i)
    for i in C:
        if i not in A and i not in B:
            s.add(i)
    return s
A = {1, 2, 3}
B = {2, 3, 4}
C = {3, 4, 5}
occur(A,B,C)
```

```python
# Find Duplicate Words in a Sentence
def dupli(n):
    s=set()
    n=n.split()
    for i in range (len(n)):
        for j in range(i+1,len(n)):
            if n[i]==n[j]:
                s.add(n[i])
    return s
n="python java python c java python"
dupli(n)
        
```

```python
#Students Who Passed All Subjects
def passed(Python,Java,DSA):
    s=set()
    for i in Python:
        if i in Java and i in DSA:
            s.add(i)
    return s
Python = {"Alice", "Bob", "Charlie", "David"}
Java = {"Bob", "Charlie", "David", "Emma"}
DSA = {"Charlie", "David", "Emma", "Frank"}
passed(Python,Java,DSA)
```

```python
#Products Available in Only One Store
def store(A,B,C):
    s=set()
    for i in A:
        if i not in B and i not in C:
            s.add(i)
    for i in B:
        if i not in A and i not in C:
            s.add(i)
    for i in C:
        if i not in A and i not in B:
            s.add(i)
    return s
A = {"Laptop", "Mouse", "Keyboard"}
B = {"Mouse", "Monitor", "Keyboard"}
C = {"Phone", "Tablet", "Monitor"}
store(A,B,C)
```

```python
#Find Missing Permissions
def missing(required,user):
    s=set()
    for i in required:
        if i not in user:
            s.add(i)
    return s

required = {"read", "write", "delete", "execute"}
user = {"read", "write"}
missing(required,user)          
```

```python
#Find Numbers Matching Multiple Conditions
#From a set of numbers, find numbers that are even AND divisible by 3.
def setnum(num):
    s=set()
    for i in num:
        if i%2==0 and i%3==0:
            s.add(i)
    return s
num = {3, 6, 9, 12, 15, 18, 20, 24}
setnum(num)
```

```python
#Find Numbers Within a Range
def numbers(num):
    s=set()
    for i in num:
        if 10<=i<=30:
            s.add(i)
    return s
num={5, 10, 15, 20, 25, 30, 35, 40}
numbers(num)
```
