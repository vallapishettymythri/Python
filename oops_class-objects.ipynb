## Class & Objects

```python
#Student details
class student:
    def __init__(self,name,age,course):
        self.name=name
        self.age=age
        self.course=course
    def display(self):
        print("name:",self.name)
        print("age:",self.age)
        print("course:",self.course)
        
st=student("Mythri",21,"AI")
st.display()
```

```python
#Car Details
class car:
    def __init__(self,brand,model,year):
        self.brand=brand
        self.model=model
        self.year=year
    def display(self):
        print(self.brand)
        print(self.model)
        print(self.year)
c=car("Maruti","Baleno","Stigma")
c.display()
```

```python
#Rectangle Area
class rectangle_area:
    def __init__(self,length,breadth):
        self.length=length
        self.breadth=breadth
    def area(self):
        print("area:",self.length*self.breadth)
a=rectangle_area(10,5)
a.area()
```

```python
#Employee details
class employee:
    def __init__(self,name,id,salary):
        self.name=name
        self.id=id
        self.salary=salary
    def display(self):
        print(self.name)
        print(self.id)
        print(self.salary)
e=employee("Mythri",101,"1c")
e.display()
```

```python
#Book Information
class book:
    def __init__(self,title,author,price):
        self.title=title
        self.author=author
        self.price=price
    def display(self):
        print(self.title)
        print(self.author)
        print(self.price)
b=book("its ends with us","ellen hover",250)
b.display()
```

```python
#Student Marks
class student:
    def __init__(self,name,eng,math,python):
        self.name=name
        self.eng=eng
        self.math=math
        self.python=python
    def display(self):
        total=self.eng+self.math+self.python
        print(self.name)
        print(self.eng)
        print(self.math)
        print(self.python)
        print("total:",total)
s=student("Mythri",80,75,90)
s.display()
```

```python
#Bank Account
class Bank_account:
    def __init__(self,holder,balance,amount):
        self.holder=holder
        self.balance=balance
        self.amount=amount
    def deposit(self):
        self.balance+=self.amount
    def display(self):
        print(self.holder)
        print(self.balance)
ba=Bank_account("Mythri",1000,500)
ba.deposit()
```

```python
ba.display()
```

```python
#Product Price
class product:
    def __init__(self,name,price):
        self.name=name
        self.price=price
    def discount(self):
        discount_amount=self.price/10
        self.price-=discount_amount
        print(self.price)
        
p=product("Mythri",50000)

p.discount()
```

```python
#Circle Calculation
class circle:
    def __init__(self,radius):
        self.radius=radius
    def area(self):
        area=3.14*(self.radius)**2
        print("area:",area)
    def circumference(self):
        circum=2*3.14*(self.radius)
        print("Circumference:",circum)
c=circle(5)
c.area()
c.circumference()
```

```python
#Mobile Phone
class mobile:
    def __init__(self,brand,model,price):
        self.brand=brand
        self.model=model
        self.price=price
    def display(self):
        print(self.brand)
        print(self.model)
        print(self.price)
m=mobile("Iphone","18 pro-max","150,000")
m.display()
```

```python
#Temperature Converter
class temperature:
    def __init__(self,celsius):
        self.celsius=celsius
    def fahre(self):
        f=self.celsius*(9/5)+32
        print(f)
t=temperature(25)
t.fahre()
```

```python
#Shopping Cart Item
class shopping:
    def __init__(self,item,price,quantity):
        self.item=item
        self.price=price
        self.quantity=quantity
    def cost(self):
        total=self.price * self.quantity
        print("item:",self.item)
        print("total:",total)
s=shopping("Notebook",50,4)
s.cost()
```

```python
#Library Book Status
class book:
    def __init__(self,title,author,is_available):
        self.title=title
        self.author=author
        self.is_available=is_available
    def borrow(self):
        self.is_available=False
        print(self.is_available)
    def display(self):
        print(self.title)
        print(self.author)
        print(self.is_available)
b=book("Its end with us", "Ellen Hoover",True)
b.borrow()
```

```python
b.display()
```

```python
#Simple Inventory
class inventory:
    def __init__(self,item,quantity,amount):
        self.item=item
        self.quantity=quantity
        self.amount=amount
    def add(self):
        self.quantity+=self.amount
        print(self.quantity)
    def remove(self):
        self.quantity-=self.amount
        print(self.quantity)

i=inventory("Laptop",10,3)
i.add()
```

```python
i.remove()
```

```python
#Movie Rating
class movie:
    def __init__(self,title,rating):
        self.title=title
        self.rating=rating
    def check_rating(self):
        if self.rating>=8:
            print("Excellent")
        elif self.rating>=6:
            print("Good")
        elif self.rating>=4:
            print("Average")
        elif self.rating<4:
            print("Poor")
m=movie("vishwanath",8.5)
m.check_rating()
```
