# Decorator

```python
#Add a message
def decorator_function(func):
    def wrapper():
        print("Before function")
        func()
    return wrapper
@decorator_function
def original():
    print("Hello")
original()
```

```python
#After function
def decorator_function(func):
    def wrapper():
        func()
        print("Function completed")
    return wrapper
@decorator_function
def original():
    print("Hello")
original()
        
```

```python
#Before + After
def decorator_function(func):
    def wrapper():
        print("Starting function")
        func()
        print("ending the function")
    return wrapper
@decorator_function
def original():
    print("Hello")
original()   
```

```python
#Create a decorator that prints "Welcome" before greet().
def decorator_function(func):
    def wrapper():
        print("Welcome")
        func()
    return wrapper
@decorator_function
def greet():
    print("Hello")
greet()
```

```python
#Write a decorator that changes this...
def deco(func):
    def wrapper():
        func()
        print("Hello")
        func()
    return wrapper
@deco
def original():
    print("********")
original() 
```
