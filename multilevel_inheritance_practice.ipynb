# Multilevel Inheritance

```python
class animal:
    def eat(self):
        print("Animals can eat")
class dog(animal):
    def bark(self):
        print("dog can bark")
class puppy(dog):
    def play(self):
        print("it can play")

a=puppy()
a.play()
```

```python
a.eat()
```

```python
class person:
    def __init__(self,name):
        self.name=name
    def display(self):
        print(self.name)
class student(person):
    def __init__(self,name,course):
        super().__init__(name)
        self.course=course
    def display1(self):
        print(self.course)
class college(student):
    def __init__(self,name,course,college):
        super().__init__(name,course)
        self.college=college
    def display2(self):
        self.display()
        self.display1()
        print(self.college)
s=college("Mythri","AIDS","St.Peter")
s.display2()
```

```python
class vehicle:
    def start(self):
        print("It starts")
class car(vehicle):
    def drive(self):
        print("Drives")
class sportscar(car):
    def race(self):
        self.start()
        self.drive()
        print("Races")
v=sportscar()
v.race()
```

```python
class person:
    def __init__(self,name):
        self.name=name
class employee(person):
    def __init__(self,name,salary):
        super().__init__(name)
        self.salary=salary
class manager(employee):
    def __init__(self,name,salary,department):
        super().__init__(name,salary)
        self.department=department
    def display(self):
        print("Name:",self.name)
        print("Salary:",self.salary)
        print("Department:",self.department)
e=manager("Mythri","1cr","CEO")
e.display()
```

```python
class product:
    def __init__(self,price):
        self.price=price
class discountprod(product):
    def prod(self):
        self.discount=self.price/10
class finalprod(discountprod):
    def display(self):
        self.prod()
        final=self.price-self.discount
        print("Original Price:",self.price)
        print("Discount:",self.discount)
        print("Final Price:",final)

p=finalprod(5000)
p.display()
```

```python
class student:
    def __init__(self,name):
        self.name=name
class marks(student):
    def __init__(self,name,marks):
        super().__init__(name)
        self.marks=marks
class result(marks):
    def __init__(self,name,marks):
        super().__init__(name,marks)
    def check_result(self):
        if self.marks>90:
            print("A")
        elif self.marks>75:
            print("B")
        elif self.marks>60:
            print("C")
        else:
            print("F")
s=result("Mythri",85)
s.check_result()
```

```python
class bankaccount:
    def __init__(self,balance,amount):
        self.balance=balance
        self.amount=amount
    def display(self):
        print("Balance",self.balance)
class savingsaccount(bankaccount):
    def deposit(self):
        self.balance+=self.amount
        self.display()
class premium(savingsaccount):
    def withdraw(self):
        self.balance-=self.amount
        self.display()
m=premium(10000,5000)
m.deposit()
```

```python
class employee:
    def __init__(self,name,salary):
        self.name=name
        self.salary=salary
class developer(employee):
    def __init__(self,name,salary,experience):
        super().__init__(name,salary)
        self.experience=experience
class seniordeve(developer):
    def calculate_bonus(self):
        if self.experience>=5:
            bonus=self.salary/20
        elif self.experience>=3:
            bonus=self.salary/15
        else:
            bonus=self.salary/10
        print(self.name)
        print(self.experience)
        print(bonus)
e=seniordeve("Mythri",60000,4)
e.calculate_bonus()
```

```python
class student:
    def __init__(self,name,rollno):
        self.name=name
        self.rollno=rollno
class marks(student):
    def __init__(self,name,rollno,python,sql,os):
        super().__init__(name,rollno)
        self.python=python
        self.sql=sql
        self.os=os
class result(marks):
    def calculate_result(self):
        total=self.python+self.sql+self.os
        avg=total/3
        if avg>=90:
            result="Excellent"
        elif avg>=75:
            result="Very Good"
        elif avg>=60:
            result="Good"
        elif avg>=40:
            result="Pass"
        else:
            result="Fail"
        print(self.name)
        print(self.rollno)
        print(total)
        print(avg)
        print(result)
s=result("Mythri",101,85,90,80)
s.calculate_result()
```
