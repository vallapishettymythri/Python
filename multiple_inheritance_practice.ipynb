# Multiple Inheritance

```python
class father:
    def father_skill(self):
        print("Driving")
class mother:
    def mother_skill(self):
        print("Cooking")
class child(father,mother):
    def child_skill(self):
        self.father_skill()
        self.mother_skill()
c=child()
c.child_skill()
```

```python
class person:
    def __init__(self,name):
        self.name=name
class college:
    def __init__(self,name,study):
        self.study=study
class student(person,college):
    def __init__(self,name,study):
        person.__init__(self,name)
        college.__init__(self,name,study)
    def display(self):
        print(self.name)
        print(self.study)
s=student("Mythri","St.Peters")
s.display()
```

```python
class calculator1:
    def __init__(self,a,b):
        self.a=a
        self.b=b
    def add(self):
        print(self.a+self.b)
class calculator2:
    def multiply(self):
        print(self.a*self.b)
class calculator(calculator1,calculator2):
    def calci(self):
        self.add()
        self.multiply()
c=calculator(10,5)
c.calci()
```

```python
class student:
    def __init__(self,python):
        self.python=python
class database:
    def __init__(self,python,sql):
        self.sql=sql
class result(student,database):
    def __init__(self,python,sql):
        student.__init__(self,python)
        database.__init__(self,python,sql)
    def res(self):
        total=self.python+self.sql
        print(total)
s=result(85,90)
s.res()
```

```python
class student:
    def __init__(self,marks):
        self.marks=marks
class attendance:
    def __init__(self,marks,attend):
        self.attend=attend

class result(student,attendance):
    def __init__(self,marks,attend):
        student.__init__(self,marks)
        attendance.__init__(self,marks,attend)
    def res(self):
        if self.marks>=40 and self.attend>=75:
            print("Eligible")
        else:
            print("Not eligible")
s=result(85,80)
s.res()
```

```python
class camera:
    def take_photo(self):
        print("Taking Photo")
class music_player:
    def play_music(self):
        print("Playing music")
class smartphone(camera,music_player):
    def make_call(self):
        self.take_photo()
        self.play_music()
        print("Making call")
c=smartphone()
c.make_call()
```

```python
class academic:
    def __init__(self,python,sql):
        self.python=python
        self.sql=sql
    def academic_total(self):
        print("Academic total:",self.python+self.sql)
class sports:
    def __init__(self,cricket,football):
        self.cricket=cricket
        self.football=football
    def sports_total(self):
        print("sports total:",self.cricket+self.football)
class student(academic,sports):
    def __init__(self,python,sql,cricket,football):
        academic.__init__(self,python,sql)
        sports.__init__(self,cricket,football)
    def overall(self):
        self.academic_total()
        self.sports_total()
        print("overall total:",self.python+self.sql+self.cricket+self.football)
s=student(85,90,70,80)
s.overall()
```

```python
class student:
    def __init__(self,name,rollno):
        self.name=name
        self.rollno=rollno
class academic:
    def __init__(self,python,sql,os):
        self.python=python
        self.sql=sql
        self.os=os
class attend:
    def __init__(self,attendance):
        self.attendance=attendance
class result(student,academic,attend):
    def __init__(self,name,rollno,python,sql,os,attendance):
        student.__init__(self,name,rollno)
        academic.__init__(self,python,sql,os)
        attend.__init__(self,attendance)
    def res(self):
        print(self.name)
        print(self.rollno)
        total=self.python+self.sql+self.os
        avg=total/3
        print("Total:",total)
        print("Average:",avg)
        print("Attendance",self.attendance)
        if avg>=75 and self.attendance>=75:
            result="Pass"
        else:
            result="Fail"
        print(result)
s=result("Mythri",101,85,90,80,82)
s.res()
```
