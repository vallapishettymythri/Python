```python
#Reverse a String
def reverse(s):
    rev=""
    for i in range (len(s)-1,-1,-1):
        rev+=s[i]
    return rev
s="hello"
reverse(s)
```

```python
#Count Vowels
def vowel(s):
    count=0
    for i in s:
        if i=='a' or i=='e' or i=='i' or i=='o' or i=='u':
            count+=1
    return count
s="programming"
vowel(s)
```

```python
#Count Consonants
def consonant(s):
    count=0
    for i in s:
        if i!='a' and i!='e' and i!='i' and i!='o' and i!='u':
            count+=1
    return count
s="python"
consonant(s)
```

```python
#Count Digits in a String
def digit(s):
    count=0
    for i in s:
        if i.isdigit():
            count+=1
    return count
s="python123"
digit(s)
```

```python
#Count Spaces
def space(s):
    count=0
    for i in s:
        if i==" ":
            count+=1
    return count
s="I love Python"
space(s)
```

```python
#Count Uppercase and Lowercase
def uclc(s):
    countuc=0
    countlc=0
    for i in s:
        if i.isupper():
            countuc+=1
        if i.islower():
            countlc+=1
    return countuc,countlc
s="PyThOn"
uclc(s)
```

```python
# Print Characters at Odd Indices
def odd(s):
    for i in range(len(s)):
        if i%2!=0:
            print(s[i])
s="python"
odd(s)
```

```python
#Remove All Spaces
def removespace(s):
    space=""
    for i in s:
        if i!=" ":
            space+=i
    return space
s="I love Python"
removespace(s)
```

```python
#Convert Lowercase to Uppercase Without upper()
def upper(s):
    res=""
    for i in s:
        var=ord(i)-32
        res+=chr(var)
    return res
s="python"
upper(s)
```

```python
#Convert Uppercase to lowercase Without lower
def lower(s):
    res=""
    for i in s:
        var=ord(i)+32
        res+=chr(var)
    return res
s="PYTHON"
lower(s)
```

```python
#Toggle Case
def toggle(s):
    res=""
    for i in s:
        if i.isupper():
            res+=i.lower()
        elif i.islower():
            res+=i.upper()
    return res
s="PyThOn"
toggle(s)
```

```python
# Find Most Frequent Character
def mostfreq(s):
    highest=0
    res=""
    for i in s:
        count=0
        for j in s:
            if i==j:
                count+=1
        if count>highest:
            highest=count
            res=i
    return res
s="banana"
mostfreq(s)
```

```python
#Find Least Frequent Character
def leastfreq(s):
    lowest = len(s)
    res = ""

    for i in s:
        count = 0

        for j in s:
            if i == j:
                count += 1

        if count < lowest:
            lowest = count
            res = i

    return res

s = "aabbccddd"
print(leastfreq(s))
```

```python
#Check Anagram- same characters, same number of each character, order can be diff
def anagram(s1,s2):
    if len(s1)!=len(s2):
        return "Not anagram"
    for i in s1:
        if s1.count(i)!=s2.count(i):
            return "Not anagram"
    return "Anagram"
        
s1="listen"
s2="silent"
anagram(s1,s2)
```

```python
#Compress a String
def compress(s):
    res=""
    done=""
    for i in range (len(s)):
        if s[i] not in done:
            count=0
            for j in range(len(s)):
                if s[i]==s[j]:
                    count+=1
            res+=s[i]
            res+=str(count)
            done+=s[i]
    return res
s="aaabbcccc"
compress(s)
```

```python
#Find Longest Consecutive Character Sequence
def consecutive(s):
    highest=0
    for i in range(len(s)):
        count=1
        for j in range (i+1,len(s)):
            if s[i]==s[j]:
                count+=1
            else:
                break
        if count>highest:
            highest=count
            result=s[i]
    return result,highest
s="aabbbccccdd"
consecutive(s)
                
```

```python
#Rotate string by k-left rotation
def rotate(s,k):
    res=""
    for i in range (k,len(s)):
        res+=s[i]
    for i in range(0,k):
        res+=s[i]
    return res
s="abcde"
k=2
rotate(s,k)
        
```

```python
#count Palindromic Substrings
def substring(s):
    count=0
    for i in range(len(s)):
        for j in range(i, len(s)):
            word=s[i:j+1]
            rev=""
            for k in range(len(word)-1,-1,-1):
                rev+=word[k]
            if word==rev:
                count+=1
    return count
s="aaa"
substring(s)    
            
```

```python
# Find Longest Palindromic Substring
def palinstring(s):
    long=""
    for i in range(len(s)):
        for j in range(i, len(s)):
            word=s[i:j+1]
            rev=""
            for k in range(len(word)-1,-1,-1):
                rev+=word[k]
            if word==rev:
                if len(word)>len(long):
                    long=word
    return long
s="babad"
palinstring(s)
                
```

```python
#Find Longest Common Prefix
def prefix(A,B,C):
    long=""
    for i in range(len(A)):
        if A[i]==B[i] and A[i]==C[i]:
            long+=A[i]
        else:
            break
    return long
A="flower"
B="flow"
C="flight"
prefix(A,B,C)
```

```python
#Check Isomorphic Strings
def isomorphic(A,B):
    if len(A)!=len(B):
        return "Not Isomorphic"
    d={}
    for i in range (len(A)):
        if A[i] in d:
            if d[A[i]]!=B[i]:
                return "Not Isomorphic"
        else:
            d[A[i]]=B[i]
    return "Isomorphic"
A = "egg"
B = "add"
isomorphic(A,B)
```
