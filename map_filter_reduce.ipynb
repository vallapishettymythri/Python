```python
#Convert List of Dates to Days
#Write a program that converts a list of date strings into the number of days since the
#Unix epoch (January 1, 1970).
from datetime import datetime
lst = ["1970-01-01", "1970-01-02", "1970-01-11"]
for x in lst:
    date=datetime.strptime(x,"%Y-%m-%d")
    epoch = datetime(1970, 1, 1)
    print((date - epoch).days)
```

```python
#Convert List of Integers to Words
#Convert a list of integers to a corresponding list of words for numbers.
def inttowords(x):
    if x==1:
        return "One"
    elif x==2:
        return "Two"
    elif x==3:
        return "Three"
    elif x==4:
        return "Four"
    elif x==5:
        return "Five"
    elif x==6:
        return "Six"
    elif x==7:
        return "Seven"
    elif x==8:
        return "Eight"
    elif x==9:
        return "Nine"
lst=[2,8,5,6]
list(map(inttowords,lst))
```

```python
#Convert a List of Time to Hours in Decimal
#Convert a list of times in &quot;HH:MM&quot; format into the number of hours in decimal.
def decimal(x):
    hours,minutes=x.split(":")
    return int(hours)+int(minutes)/60
lst=["02:30", "05:45", "10:15", "12:30"]
list(map(decimal,lst))
```

```python
#Convert List of Coordinates to Distances from Origin
#Convert a list of coordinates [(x1, y1), (x2, y2)] into the Euclidean distance
#from the origin (0, 0).
import math
def distance(point):
    x,y=point
    return math.sqrt(x**2+y**2)
lst = [(3,4), (5,12), (8,15)]
print(list(map(distance, lst)))
```

### filter()

```python
#Filter Even Numbers
#Write a program to filter even numbers from a list of integers.
def even(x):
    return x%2==0
lst=[2,3,4,5,6]
list(filter(even,lst))
```

```python
#Filter Words Longer than 5 Characters
#Write a program to filter out all words that are longer than 5 characters.
def length(x):
    return len(x)>5
lst=["Hi","Python","Monkey","Peacock"]
list(filter(length,lst))
```

```python
#Filter Positive Numbers
#Write a program that filters only the positive numbers from a list of integers.
def positive(x):
    return x>0
lst=[-1,3,8,-10,9]
list(filter(positive,lst))
```

```python
#Filter Odd Numbers
#Write a program to filter all odd numbers from a list of integers. 
def odd_num(x):
    return x%2!=0
lst=[1,2,3,4,5,6]
list(filter(odd_num,lst))
```

```python
#Filter Palindromes
#Write a program to filter palindromic words from a list of strings.
def palindrome(x):
    return x==x[::-1]
lst=["madam", "hello", "level", "python", "radar"]
list(filter(palindrome,lst))
```

```python
#Filter Prime Numbers
#Write a program that filters all prime numbers from a list of integers.
def prime(x):
    if x < 2:
        return False

    for i in range(2, x):
        if x % i == 0:
            return False

    return True

lst = [2, 4, 5, 7, 9, 11, 15, 17]

print(list(filter(prime, lst)))
```

```python
#Filter Non-Empty Strings
#Write a program to filter out all empty strings from a list of strings.
def empty_strings(x):
    if x!="":
        return x
lst=["Python", "", "Hello", "", "World"]
list(filter(empty_strings,lst))
```

### reduce()

```python
from functools import reduce
```

```python
#Find the Sum of a List of Numbers
#Write a program to find the sum of all elements in a list.
def sum(a,b):
    return a+b
lst=[1,2,3,4,5]
reduce(sum,lst)
```

```python
#Find the Maximum of a List of Numbers
#Write a program to find the maximum number from a list of numbers.
def maximum(a,b):
    if a>b:
        return a
    else:
        return b
lst=[6,1,9,2,3]
reduce(maximum,lst)
```

```python
#Find the Minimum of a List of Numbers
#Write a program to find the minimum number from a list of numbers.
def minimum(a,b):
    if a<b:
        return a
    else:
        return b
lst=[6,1,9,2,3]
reduce(minimum,lst)
```

```python
#Concatenate a List of Strings
#Write a program to concatenate all strings in a list into a single string.
def conacatenate(a,b):
    return a+b
lst=["Hello", " ", "World", "!"]
reduce(conacatenate,lst)
```

```python
#Find the Product of All Numbers
#Write a program to find the product of all elements in a list.
def product(a,b):
    return a*b
lst=[1,2,3,4,5]
reduce(product,lst)
```

```python
#Find the Average of a List of Numbers
#Write a program to find the average of all numbers in a list.
def add(a,b):
    return a+b

lst = [1, 2, 3, 4, 5]
total=reduce(add,lst)
average=total/len(lst)
print(average)
```

```python
#Count the Total Number of Characters in a List of Strings
#Write a program to count the total number of characters in a list of strings.
def length(x):
    return len(x)
def add(a,b):
    return a+b
lst=["Hello", "Python", "World"]
lengths=list(map(length,lst))
print(reduce(add,lengths))
```

```python
#Find the Longest Word
#Write a program to find the longest word in a list of strings.
def longest(a,b):
    if len(a)>len(b):
        return a
    else:
        return b
lst=["cat", "elephant", "dog", "butterfly"]
print(reduce(longest,lst))
```

```python
#Sum the Squares of the Numbers
#Write a program to sum the squares of all the numbers in a list.
def squaresum(a,b):
    return a+b**2
lst=[1, 2, 3, 4, 5]
reduce(squaresum,lst)
```

```python
#Combine Two Lists into a Tuple
#Write a program that combines two lists of equal length into a list of tuples.
list1 = [1, 2, 3]
list2 = ["a", "b", "c"]

def combine(result,i):
    result.append((list1[i],list2[i]))
    return result

reduce(combine,range(len(list1)),[])
```

```python
#Flatten a List of Lists
#Write a program to flatten a list of lists into a single list.
lst=[[1, 2], [3, 4], [5, 6]]
def flatten(result,i):
    for x in lst[i]:
        result.append(x)
    return result
reduce(flatten,range(len(lst)),[])
```

```python
#Find the Factorial of a List of Numbers
#Write a program to find the factorial of each number in the list and reduce them to a
#total product.
lst=[2, 3, 4]

def multiply(a,b):
    return a*b
def factorial(n):
    return reduce(multiply,range(1,n+1),1)
def product(a,b):
    return a*factorial(b)
reduce(product,lst,1)
```

```python
#Count the Total Number of Digits in All Numbers
#Write a program to count the total number of digits in a list of numbers.
lst=[123, 45, 6789, 10]
def digits(a):
    return len(str(a))
def add(a,b):
    return a+b
counts=list(map(digits,lst))
reduce(add,counts)
```

```python
#Find the Mode of a List of Numbers
#Write a program to find the mode (most frequent element) of a list of numbers.
lst=[1, 2, 2, 3, 4, 2, 5]
def mode(a,b):
    if lst.count(a)>lst.count(b):
        return a
    else:
        return b

reduce(mode,lst)
```

```python
#Reduce List to a Single Character String
#Write a program to reduce a list of characters to a single string using concatenation.
lst=['P', 'y', 't', 'h', 'o', 'n']
def concatenate(a,b):
    return a+b
reduce(concatenate,lst)
```

```python
#Find the Sum of Factorials of Numbers in List
#Write a program to find the sum of the factorials of numbers in a list.
def multiply(a,b):
    return a*b
def factorial(n):
    return reduce(multiply,range(1,n+1),1)
def add(a,b):
    return a+factorial(b)
lst=[2, 3, 4]
reduce(add,lst,0)
```

```python
#Combine Lists of Numbers into a Single List by Arithmetic Operations
#Write a program to add corresponding elements of two lists of numbers.
lst1=[1, 2, 3]
lst2=[4, 5, 6]
def add(result,i):
    result.append(lst1[i]+lst2[i])
    return result
reduce(add,range(len(lst1)),[])
```
