# Hybrid inheritance 

```python
class animal:
    def eat(self):
        print("Eating")
class dog(animal):
    def bark(self):
        print("Barking")
class cat(animal):
    def meow(self):
        print("Meowing")
class pet(dog,cat):
    def play(self):
        print("playing")

p=pet()
p.play()
p.eat()
p.meow()
p.bark()
```

```python
class person:
    def __init__(self,name):
        self.name=name

class student(person):
    def __init__(self,name,course):
        person.__init__(self,name)
        self.course=course

class employee(person):
    def __init__(self,name,salary):
        person.__init__(self,name)
        self.salary=salary

class manager(student,employee):
    def __init__(self,name,course,salary):
        student.__init__(self,name,course)
        employee.__init__(self,name,salary)

    def display(self):
        print("Name:",self.name)
        print("Course:",self.course)
        print("Salary:",self.salary)

m=manager("Mythri","AI & DS",60000)
m.display()
```

```python
class person:
    def __init__(self,name):
        self.name=name
class student(person):
    def __init__(self,name,python,sql):
        person.__init__(self,name)
        self.python=python
        self.sql=sql
    def academic_total(self):
        self.total=self.python+self.sql
        print(self.total)
class employee(person):
    def __init__(self,name,salary):
        person.__init__(self,name)
        self.salary=salary
    def calculate_bonus(self):
        bonus=self.salary/10
        print(bonus)
class result(student,employee):
    def __init__(self,name,python,sql,salary):
        student.__init__(self,name,python,sql)
        employee.__init__(self,name,salary)
    def display(self):
        print(self.name)
        self.academic_total()
        print(self.salary)
        self.calculate_bonus()
        if self.total>=150:
            print("Eligible")
        else:
            print("Not eligible")
```

```python
r=result("Mythri",85,90,60000)
r.display()
```
