```python
lst = ["Hyderabad", "Delhi", "Mumbai", "Delhi", "Pune", "Goa"]
def no_repeat(lst):
    high=0
    for i in range(len(lst)):
        ans=[]
        for j in range(i,len(lst)):
            if lst[j] not in ans:
                ans.append(lst[j])
            else:
                break
        if len(ans)>high:
            high=len(ans)
            
    return high
no_repeat(lst)
```

```python
lst = [
    [101, 102, 103],
    [104, 105, 102],
    [106, 107, 108]
]
def repeated(lst):
    for i in range(len(lst)):
        for j in range(i+1,len(lst)):
            for x in lst[i]:
                if x in lst[j]:
                    return x
repeated(lst)   
```
