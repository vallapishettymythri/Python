```python
#Sort Dictionary by Value
#Write a Python script to sort (ascending and descending) a dictionary by value.
def order(marks):
    keys = list(marks.keys())

    for i in range(len(keys)):
        for j in range(i + 1, len(keys)):
            if marks[keys[i]] > marks[keys[j]]:
                keys[i], keys[j] = keys[j], keys[i]

    ascending = {}
    descending = {}

    for i in keys:
        ascending[i] = marks[i]

    for i in keys[::-1]:
        descending[i] = marks[i]

    print("Ascending:", ascending)
    print("Descending:", descending)


marks = {
    "Math": 85,
    "Python": 92,
    "C": 78,
    "Java": 88
}

order(marks)
```

```python
#Concatenate Dictionaries
#Write a Python script to concatenate the following dictionaries to create a new one.
def concatenate(dic1,dic2,dic3):
    result={}
    for key,value in dic1.items():
        result[key]=value
    for key,value in dic2.items():
        result[key]=value
    for key,value in dic3.items():
        result[key]=value
    return result
dic1={1:10, 2:20}
dic2={3:30, 4:40}
dic3={5:50,6:60}
concatenate(dic1,dic2,dic3)
```

```python
#Get Maximum and Minimum Values of a Dictionary
#Write a Python program to get the maximum and minimum values of a
#dictionary.
def minmax(d):
    values=list(d.values())
    high=values[0]
    low=values[0]
    for i in values:
        if i>high:
            high=i
        if i<low:
            low=i
    return high,low
d = {1: 10, 2: 50, 3: 20, 4: 5}
minmax(d)
```

```python
#Combine Two Dictionaries by Adding Values
#Write a Python program to combine two dictionary by adding values for
#common keys.
def add(d1,d2):
    result={}
    for key, value in d1.items():
        result[key]=value
    for key,value in d2.items():
        if key in result:
            result[key]=result[key]+value
        else:
            result[key]=value
    return result
d1={'a':100,'b':200,'c':300}
d2={'a':300,'b':200,'d':400}
add(d1,d2)
```

```python
#Sort Counter by Value
#Write a Python program to sort Counter by value.
def sortvalue(d):
    result=list(d.items())
    for i in range (len(result)):
        for j in range(i+1,len(result)):
            if result[i][1]<result[j][1]:
                result[i],result[j]=result[j],result[i]
    return result
d = {'Math': 81, 'Physics': 83, 'Chemistry': 87}
sortvalue(d)
```

```python
#Create Dictionary from Two Lists Without Losing Duplicates
#Write a Python program to create a dictionary from two lists without losing
#duplicate values.
def dict(lst1,lst2):
    result={}
    for i in range (len(lst1)):
        if lst1[i] not in result:
            result[lst1[i]]=[]
        result[lst1[i]].append(lst2[i])
    return result
lst1 = ['Class-V', 'Class-VI', 'Class-VII', 'Class-VIII']
lst2 = [1, 2, 2, 3]
dict(lst1,lst2)
```

```python
#Replace Dictionary Values with Their Sums
#Write a Python program to replace dictionary values with their sums.
def sums(d):
    result={}
    for key,value in d.items():
        total=0
        for i in value:
            total+=i
        result[key]=total
    return result

d = {
    'a': [10, 20, 30],
    'b': [5, 15],
    'c': [2, 3, 4, 5]
}
sums(d)
        
```

```python
#Convert Multiple Lists to a Nested Dictionary
#Write a Python program to convert more than one list to a nested dictionary.
def nested(lst1,lst2,lst3):
    result=[]
    for i in range(len(lst1)):
        d={}
        d[lst1[i]]={lst2[i]:lst3[i]}
        result.append(d)
    return result
lst1 = ['S001', 'S002', 'S003', 'S004']
lst2 = ['Adina Park', 'Leyton Marsh', 'Duncan Boyle', 'Saim Richards']
lst3 = [85, 98, 89, 92]
nested(lst1,lst2,lst3)
```

```python
#Filter Students by Height and Weight
#Write a Python program to filter the height and width of students, which are
#stored in a dictionary.
def personality(d):
    result={}
    for i,j in d.items():
        for height,weight in [j]:
            if height >=6 and weight>=70:
                result[i]=j
    return result
d = {
    'Cierra Vega': (6.2, 70),
    'Alden Cantrell': (5.9, 65),
    'Kierra Gentry': (6.0, 68),
    'Pierre Cox': (5.8, 66)
}
personality(d)
```

```python
#Group Key-Value Pairs into Dictionary of Lists
#Write a Python program to create a dictionary grouping a sequence of key-
#value pairs into a dictionary of lists.
def pairs(d):
    result={
        'yellow':[],
        'blue':[],
        'red':[]
    }
    for i,j in d:
        result[i].append(j)
    return result
d = [
    ('yellow', 1),
    ('blue', 2),
    ('yellow', 3),
    ('blue', 4),
    ('red', 1)
]
pairs(d)
```

```python
#Split Dictionary of Lists into List of Dictionaries
#Write a Python program to split a given dictionary of lists into lists of
#dictionaries.
def splitdict(d):
    result=[]
    for i in range(len(d['Science'])):
        temp={}
        temp['Science']=d['Science'][i]
        temp['Language']=d['Language'][i]
        result.append(temp)
    return result
d = {
    'Science': [88, 89, 62, 95],
    'Language': [77, 78, 84, 80]
}
splitdict(d)
```

```python
#Convert String Values in Dictionary to Numeric Types
#Write a Python program to convert string values of a given dictionary into
#integer/float datatypes.
def convert(d):
    result=[]
    for i in d:
        temp={}
        for key, value in i.items():
            temp[key]=int(value)
        result.append(temp)

    return result
d = [
    {'x': '10', 'y': '20', 'z': '30'},
    {'p': '40', 'q': '50', 'r': '60'}
]
convert(d)
```
