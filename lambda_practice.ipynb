#### Dictionary Operations

```python
#Sort a Dictionary by Keys
#Write a lambda function that sorts a dictionary by its keys.
sorting=lambda d: dict(sorted(d.items(),key=lambda x:x[0]))
sorting({"b":2,"a":1,"c":3})
```

```python
#Sort a Dictionary by Values
#Write a lambda function that sorts a dictionary by its values.
sorting=lambda d: dict(sorted(d.items(),key=lambda x:x[1]))
sorting({"b":2,"a":1,"c":3})
```

```python
#Find the Key with Maximum Value
#Write a lambda function that finds the key with the maximum value in a
#dictionary.
maximum=lambda d:max(d.values())
maximum({"a":10,"b":25,"c":15})
```

```python
#Find the Key with Minimum Value
#Write a lambda function that finds the key with the minimum value in a
#dictionary.
minimum=lambda d: min(d.values())
minimum({"a":10,"b":25,"c":15})
```

```python
#Extract All Values from a Dictionary
#Write a lambda function that extracts all the values from a dictionary into a
#list.
values=lambda d: d.values()
list(values({"a": 10, "b": 20, "c": 30}))
```

```python
#Filter a Dictionary by Value
# Write a lambda function that filters a dictionary, keeping only keys with
#values greater than a specified number.
greater=lambda d:dict(filter(lambda x: x[1]>20,d.items()))
print(greater({"a":10,"b":25,"c":15,"d":30}))
```

```python
#Invert a Dictionary (Keys become Values and Values become Keys)
# Write a lambda function that inverts the keys and values of a dictionary.
inverse=lambda d: dict(map(lambda x:(x[1],x[0]),d.items()))
inverse({"a": 1, "b": 2, "c": 3})
```

```python
#Add a Prefix to All Keys in a Dictionary
#Write a lambda function that adds a prefix to all the keys in a dictionary.
prefix=lambda d: dict(map(lambda x: ("pre_"+x[0],x[1]),d.items()))
prefix({"a": 10, "b": 20, "c": 30})
```

```python
#Update Values Based on a Condition
#Write a lambda function that increases the value of keys in a dictionary by 10
#if the value is greater than 50.
values=lambda d: dict(map(lambda x: (x[0],x[1]+10) if x[1]>50 else (x[0],x[1]),d.items()))
values({"a": 40, "b": 60, "c": 80, "d": 30})
```

```python
#Create a Dictionary from Two Lists
#Write a lambda function that combines two lists into a dictionary where the
#first list contains the keys and the second list contains the values.
create=lambda keys,values: dict(map(lambda i: (keys[i],values[i]),range(len(keys))))
print(create([1, 2, 3], ["a", "b", "c"]))
```

```python
#Write a Python program to find the second lowest total marks of any
#student(s) from the given names and marks of each student using lists and
#lambda. Input the number of students, the names and grades of each student.
students = [
    ["S ROY", 1],
    ["B BOSE", 3],
    ["N KAR", 2],
    ["C DUTTA", 1],
    ["G GHOSH", 1]
]
grades=list(map(lambda x: x[1],students))
grades=sorted(set(grades))
second_lowest=grades[1]
for student in students:
    if student[1]==second_lowest:
        print(student[0])
```
