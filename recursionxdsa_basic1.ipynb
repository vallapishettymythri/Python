```python
#Check whether a string is a palindrome.
def recursive_palindrome(s):
    if len(s)<=1:
        return True
    else:
        if s[0] !=s[-1]:
            return False
        else:
            return recursive_palindrome(s[1:-1])
s = "madam"
recursive_palindrome(s)
```

```python
#Replace all occurrences of a character with another character.
def recursive_occur_char(s,old_char,new_char):
    if s=="":
        return ""
    else:
        if s[0]==old_char:
            return new_char+recursive_occur_char(s[1:],old_char,new_char)
        else:
            return s[0]+recursive_occur_char(s[1:],old_char,new_char)
s = "banana"
old_char = "a"
new_char = "o"
recursive_occur_char(s,old_char,new_char)
```

```python
#Check whether two strings are equal without using == for the complete strings.
def recursive_equal(s1, s2):
    if s1 == "" and s2 == "":
        return True
    elif s1 == "" or s2 == "":
        return False
    else:
        if s1[0] != s2[0]:
            return False
        else:
            return recursive_equal(s1[1:], s2[1:])
s1 = "python"
s2 = "python"
recursive_equal(s1, s2)
```

```python
#Calculate the sum of all numeric values in a dictionary.
def recursive_numeric(d):
    values=list(d.values())
    if values==[]:
        return 0
    else:
        return values[0]+recursive_numeric({key:value for key,value in list(d.items())[1:]})
d={"a": 10, "b": 20, "c": 30, "d": 40}
print(recursive_numeric(d))
```

```python
#Find the maximum value in a dictionary.
def recursive_max(d):
    values=list(d.values())
    if len(values)==1:
        return values[0]
    else:
        remaining=recursive_max({key:value for key,value in list(d.items())[1:]})

        if values[0]>remaining:
            return values[0]
        else:
            return remaining

d = {"a": 10, "b": 50, "c": 30, "d": 40}
recursive_max(d)
```

```python
#Search whether a particular key exists.
def recursive_search_key(d,target):
    keys=list(d.keys())
    if keys==[]:
        return False
    else:
        if keys[0]==target:
            return True
        else:
            return recursive_search_key({key:value for key,value in list(d.items())[1:]},target)
d = {"a": 10, "b": 20, "c": 30, "d": 40}
target = "c"
recursive_search_key(d,target)
```

```python
#Find the value associated with a given key.
def recursive_associated(d,target):
    lst=list(d.items())
    if lst==[]:
        return None
    else:
        if lst[0][0]==target:
            return lst[0][1]
        else:
            return recursive_associated({key:value for key,value in lst[1:]},target)
d = {"a": 10, "b": 20, "c": 30, "d": 40}
target = "c"
recursive_associated(d,target)
```

```python
#Merge two dictionaries recursively.
def recursive_merge(d1,d2):
    items=list(d2.items())

    if items==[]:
        return d1
    else:
        d1[items[0][0]]=items[0][1]
        return recursive_merge(d1,{key:value for key ,value in items[1:]})
d1 = {"a": 10, "b": 20}
d2 = {"c": 30, "d": 40}

recursive_merge(d1, d2)
```
