```python
def subarray(lst,k):
    longest=0
    for i in range(len(lst)):
        sum=0
        for j in range(i,len(lst)):
            sum+=lst[j]

            if sum==k:
                length=j-i+1

                if length>longest:
                    longest=length
    return longest
lst=[1, 2, 3, 1, 1, 1, 5]
k=6
subarray(lst,k)
```

```python
lst = [2, 7, 11, 15]
target = 9
def pairsum(lst,target):
    sum=0
    for i in range(len(lst)):
        for j in range(i+1,len(lst)):
            sum=lst[i]+lst[j]

            if sum==target:
                return lst[i],lst[j]
pairsum(lst,target)
```

```python
def pairsum(lst,target):
    dict={}
    for i in range(len(lst)):
        needed=target-lst[i]
        if needed in dict:
            return dict[needed],i
        dict[lst[i]]=i
pairsum(lst,target) 
```

```python
def repeating(lst):
    dict={}
    for i in range(len(lst)):
        if lst[i] not in dict:
            dict[lst[i]]=1
        else:
            return lst[i]
lst = [5, 3, 8, 3, 2, 5]
repeating(lst)
```

```python
def freq(lst):
    high=0
    n=0
    for i in range (len(lst)):
        count=0
        for j in range(len(lst)):
            if lst[i]==lst[j]:
                count+=1

                if count> high:
                    high=count
                    n=lst[i]
    return n,high
lst = [4, 2, 4, 3, 2, 4, 5]
freq(lst)
```

```python
def two_list(lst1,lst2):
    if len(lst1)!=len(lst2):
        return False
    for i in range(len(lst1)):
        if lst1.count(lst1[i])!= lst2.count(lst1[i]):
            return False
    return True

lst1 = [1, 2, 2, 3, 4]
lst2 = [2, 1, 4, 2, 3]  
two_list(lst1,lst2)
```

```python
def encouter(n):
    high=0
    while n>0:
        digit=n%10
        if digit> high:
            high=digit
        n//=10
    return high
n=58394
encouter(n)
```

```python
def occur(lst):
    for i in range(len(lst)):
        count=0
        for j in range(len(lst)):
            if lst[i]==lst[j]:
                count+=1
        if count==1:
            print(lst[i])
lst = [4, 2, 4, 7, 2, 9, 5]
occur(lst)
```

```python
def compressed(s):
    com=""
    visited=[]
    for i in range (len(s)):
        if s[i] not in visited:
            count=0
            for j in range (len(s)):
                if s[i]==s[j]:
                    count+=1
            com+=s[i]+str(count)
            visited.append(s[i])
    return com
s = "aaabbccccd"
compressed(s)            
                
```

```python
marks = {
    "Ravi": 78,
    "Anu": 92,
    "Sita": 85,
    "Kiran": 88
}
def highest_score(marks):
    high=0
    s=""
    for i,j in marks.items():
        if j>high:
            high=j
            s=i
    return high,s
highest_score(marks)
```

```python
lst = [
    [1, 2, 3],
    [4, 5],
    [6, 7, 8]
]
def sum_list(lst):
    sum=0
    for i in lst:
        for j in i:
            sum+=j
    return sum
sum_list(lst)
```

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
def diagonal(matrix):
    sum=0
    for i in range (len(matrix)):
        sum+=matrix[i][i]
    return sum
diagonal(matrix)
```

```python
marks = {
    "Ravi": 65,
    "Anu": 90,
    "Sita": 75,
    "Kiran": 50
}

limit = 70
def greater(marks,limit):
    dic={}
    for i,j in marks.items():
        if j>limit:
            dic[i]=j
    return dic
greater(marks,limit)
```
