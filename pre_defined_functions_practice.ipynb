```python
#count how many values are integers.
values = [10, "hello", 20, "Python", 30]
count=0
for i in values:
    if isinstance(i,int):
        count+=1
print(count)
```

### Filter()

```python
#Create a function even(x) and use filter() to get only the even numbers.
numbers = [1, 2, 3, 4, 5, 6]
def even(x):
    return x%2==0
result=filter(even,numbers)
print(list(result))
```

```python
#Create a function positive(x) and use filter() to select only the positive
#numbers
numbers = [-10, 5, -3, 8, -2, 15]
def pos(x):
    return x>0
list(filter(pos,numbers))
```

```python
#Create a function that selects only the words whose length is greater than 4.
words = ["python", "java", "sql", "javascript", "go"]
def length(x):
    return len(x)>4
list(filter(length,words))
```

```python
#Create a function that selects only the numbers greater than 10 using filter().
num = [12, 7, 20, 3, 15, 8]
def greater(x):
    return x>10
list(filter(greater,num))
```

```python
#Create a function that selects only the numbers divisible by 5 using filter().
numbers = [10, 15, 22, 31, 40, 53]
def divisible(x):
    return x%5==0
list(filter(divisible,numbers))
```

### reduce()

```python
#Use reduce() to find the sum of all numbers.
from functools import reduce
numbers = [2, 4, 6, 8]
def add(a,b):
    return a+b
reduce(add,numbers)
```

```python
#Use reduce() to find the product of all numbers.
from functools import reduce
numbers = [2, 3, 4, 5]
def multiply(a,b):
    return a*b
reduce(multiply,numbers)
```

```python
#Use reduce() to find the largest number.
from functools import reduce
numbers = [10, 25, 7, 40, 15]
def largest(a,b):
    if a>b:
        return a
    else:
        return b
reduce(largest,numbers)
```

```python
#Use reduce() to find the smallest number.
from functools import reduce
numbers = [5, 12, 8, 20, 3]
def smallest(a,b):
    if a<b:
        return a
    else:
        return b
reduce(smallest,numbers)
```

```python
#Use reduce() to calculate the difference by subtracting each next number from the accumulated result.
from functools import reduce
numbers = [2, 4, 6, 8]
def difference(a,b):
    return a-b
reduce(difference,numbers)
```

### enumerate()

```python
names = ["Alice", "Bob", "Charlie", "David"]
for i,j in enumerate(names):
    print(i,j)
```

```python
#Use enumerate() to print the subject number and mark, starting the subject number from 1.
marks = [85, 72, 91, 68]
for i,j in enumerate(marks,start=1):
    print(i,j)
```

```python
#Use enumerate() to print only the index of numbers that are greater than 30.
numbers = [10, 25, 30, 45, 50]
for i,j in enumerate(numbers):
    if j>30:
        print(i)
```

```python
#Use enumerate() to print only the names whose index is even.
names = ["Alice", "Bob", "Charlie", "David"]
def even(names):
    for i,j in enumerate(names):
        if i%2==0:
            print(j)

even(names)
```

```python
#Use enumerate() to find and print the value and index of the largest number.
numbers = [10, 20, 30, 40, 50]
def largest(numbers):
    large=numbers[0]
    index=0
    for i,j in enumerate(numbers):
        if j>large:
            large=j
            index=i
    print(index,large)
numbers = [10, 20, 30, 40, 50]
largest(numbers)
```

### lambda()

```python
#Create a lambda function that takes a number and returns its square.
square=lambda x:x*x
print(square(5))
```

```python
#Use lambda and map() to create a new list containing the squares
#of all numbers.
numbers = [1, 2, 3, 4, 5]
result=list(map(lambda x:x*x, numbers))
print(result)
```

```python
#Use lambda and filter() to select only the even numbers.
numbers = [10, 15, 22, 31, 40, 53]
result=list(filter(lambda x: x%2==0, numbers))
print(result)
```

```python
#Use lambda and map() to convert every word to uppercase.
words = ["python", "java", "sql"]
result=list(map(lambda x: x.upper(), words))
print(result)
```

```python
#Use lambda and filter() to select numbers greater than 10 and even.
numbers = [5, 12, 7, 20, 3, 18]
result=list(filter(lambda x: x>10 and x%2==0, numbers))
print(result)
```
