```python
# Write a Python program to add an item to a tuple.
def add(tpl,item):
    new=[]
    for i in tpl:
        new.append(i)
    new.append(item)
    return tuple(new)
    
tpl=("Mythri",21,"Hyderbad")
item="AIDS"
add(tpl,item)
```

```python
#Write a Python program to find repeated items in a tuple.
def repeat(tpl):
    new=[]
    for i in tpl:
        count=0
        for j in tpl:
            if i==j:
                count+=1
        if count>1 and i not in new:
            print(i)
            new.append(i)
tpl = (10, 20, 30, 20, 40, 10, 50, 20)
repeat(tpl)
```

```python
#Write a Python program to check whether an element exists within a tuple.
def exist(tpl,item):
    for i in tpl:
        if i==item:
            return "Element exists"
    return "Element not exists"
tpl = (10, 20, 30, 40, 50)
item = 30
exist(tpl,item)
```

```python
#Write a Python program to remove an item from a tuple.
def remove(tpl,item):
    new=[]
    for i in tpl:
        if i!=item:
            new.append(i)
    return new
tpl = (10, 20, 30, 40, 50)
item = 30
remove(tpl,item)
```

```python
#Write a Python program to find the index of an item in a tuple.
def index(tpl,item):
    for i in range (len(tpl)):
        if tpl[i]==item:
            return i
tpl=(10,20,30,40,50,60)
item=30
index(tpl,item)
```

```python
#Write a Python program to find the length of a tuple.
def length(tpl):
    count=0
    for i in tpl:
        count+=1
    return count
tpl=(10,20,30,40,50,60)
length(tpl)
```
