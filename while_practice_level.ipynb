## level 1

```python
#Print all even numbers from 1 to 20.
i=1
while i<=20:
    if i%2==0:
        print(i)
    i+=1
```

```python
#Find the second largest digit.
n=int(input("n:"))
large=-1
second=-1
while n>0:
    digit=n%10
    if(digit > large):
        second=large
        large=digit
    elif digit > second and digit !=large:
        second=digit
    n//=10
print(large,second)

    
    
```

```python
#Find the second smallest digit.
n=int(input("n:"))
small=10
second=10
while n>0:
    digit=n%10
    if(digit < small):
        second=small
        small=digit
    elif digit < second and digit !=small:
        second=digit
    n//=10
print(small,second)
```

```python
#Check whether a number is a Strong number.
n=int(input("n:"))
sum=0
temp=n
while temp>0:
    digit=temp%10
    product=1
    i=1
    while i<=digit:
        product*=i
        i+=1
    sum+=product
    temp//=10
if sum==n:
    print(sum)
else:
    print("not strong number")
```

```python
#Check whether all digits are the same.
n=int(input("n:"))
num=n%10
flag=True
while n>0:
    digit=n%10
    if digit!=num:
        flag=False
        break
    n//=10
if flag:
    print("same")
else:
    print("not same")
```

## Level 4- Logic building

```python
#Count how many times a particular digit occurs.
n=int(input("n:"))
digit=int(input("digit:"))
count=0
while n>0:
    digits=n%10
    if digits==digit:
        count+=1
    n//=10
print(count)
        
```

```python
#Remove the last digit.
n=int(input("n:"))
last=n//10
print(last)
    
```

```python
#Remove the first digit
n=int(input("n:"))
count=0
temp=n
while temp>0:
    count+=1
    temp//=10
tens=10**(count-1)
answer=n%tens
print(answer)
    
```

```python
#Find the middle digit (odd number of digits).
n=int(input("n:"))
count =0
temp=n
while temp>0:
    count+=1
    temp//=10
middle=count//2
while middle > 0:
    n //= 10
    middle -= 1
print(n%10)
```

```python
#Swap the first and last digit.
n=int(input("n:"))
last=n%10
count=0
temp=n
while temp>0:
    count+=1
    temp//=10
tens=10**(count-1)
first=n//tens
middle = (n % tens)//10
print(last*tens+middle*10+first)
```

```python
#Check whether a number contains the digit 5.
n=int(input("n:"))
num=5
flag=False
while n>0:
    digit=n%10
    if digit==num:
        flag=True
        break
    n//=10
if flag:
    print("5 is present")
else:
    print("5 is not present")

```

```python
#Find the average of all digits.
n=int(input("n:"))
count=0
sum=0
temp=n
while temp>0:
    count+=1
    temp//=10
while n>0:
    digit=n%10
    sum+=digit
    n//=10
average=sum/count
print(average)
```

```python
#print the digits in reverse order.
n=int(input("n:"))
count=0
reverse=0
while n>0:
    digit=n%10
    reverse = reverse * 10 + digit
    n//=10
print(reverse)
    
    
    
    
```

```python
#print the digits in reverse order. One digit per line.
n = int(input("n:"))

while n > 0:
    digit = n % 10
    print(digit)
    n //= 10
```

```python
#Print each digit on a new line from left to right.
n=int(input("n:"))
count=0
temp=n
while temp >0:
    count+=1
    temp//=10
tens=10**(count-1)
while n>0:
    digit=n//tens
    print(digit)
    n %= tens
    tens //= 10
    
```
