```python
#Count how many 1 bits are present in a positive integer using bitwise operations.
n=int(input("n:"))
count=0
while n>0:
    if n&1:
        count+=1
    n=n>>1
print(count)
```

## level 3- Problem solving

```python
#Clear the rightmost set bit of a number.
n=int(input("n:"))
result= n&(n-1)
print(result)
```

```python
#Check whether the nth bit of a number is set.
num=int(input("num:"))
n=int(input("n:"))
if num&(1<<n):
    print("Set")
else:
    print("not set")
```

```python
#Set the nth bit of a number.
num=int(input("num:"))
n=int(input("n:"))
result=num|(1<<n)
print(result)
```

```python
#Clear the nth bit of a number.
num=int(input("num:"))
n=int(input("n:"))
result=num&~(1<<n)
print(result)
```

```python
#Toggle the nth bit of a number.
num=int(input("num:"))
n=int(input("n:"))
result=num^(1<<n)
print(result)
```

```python
#Given a number, find the number obtained after reversing its bits within a fixed 8-bit representation.
n=int(input("n:"))
result=0
for i in range(8):
    bit=n&1
    result=result|(bit << (7-i))
    n=n>>1
print(result)
```

```python
#Given two integers, find how many bits are different between them.
a=int(input("a:"))
b=int(input("b:"))
count=0
result=a^b
while result>0:
    if result & 1:
        count+=1
    result=result>>1
print(count)
```

```python
#Find the XOR of all numbers from 1 to n.
n=int(input("n:"))
result=0
for i in range(1,n+1):
    result=result^i
print(result)
```

## Level-4 Challenging bit manipulation

```python
#Check whether two integers have opposite signs using bitwise operations.
a=int(input("a:"))
b=int(input("b:"))
if (a^b)<0:
    print("opposite signs")
else:
    print("both are same signs")
```

```python
#Find the next power of 2 greater than or equal to a given positive integer.
n=int(input("n:"))
power=1
while power<n:
    power=power<<1
print(power)
```

```python
#Find the position of the rightmost set bit.
n=int(input("n:"))
position=0
while(n&1)==0:
    n=n>>1
    position+=1
print(position)
```

```python
#Toggle all bits of it after the most significant bit including the most significant bit.
n = int(input("n: "))

mask = 1

while mask <= n:
    mask = mask << 1

mask = mask - 1

result = n ^ mask

print(result)
```
