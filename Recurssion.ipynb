### Level 1- Basic recursion

```python
#1.Write a recursive function fun(n) to print numbers from n down to 1.
def fun(n):
    if n==0:
        return
    print(n)
    fun(n-1)
n=int(input("n:"))
fun(n)
```

```python
#2. Write a recursive function fun(n) to print numbers from 1 to n.
def fun(n):
    if n==0:
        return
    fun(n-1)
    print(n)
n=int(input("n:"))
fun(n)
```

```python
#3. Write a recursive function fun(n) to print "Hello" exactly n times.
def fun(n):
    if n==0:
        return 
    fun(n-1)
    print("hello")
n=int(input("n:"))
fun(n)
```

```python
#4. Write a recursive function fun(n) to find the sum of numbers from 1 to n.
def sum_num(n):
    if n==0:
        return 0
    else:
        return n+sum_num(n-1)
n=int(input("n:"))
print(sum_num(n))
        
```

```python
#5. Write a recursive function fun(n) to find the factorial of n.
def fact(n):
    if n==1:
        return 1
    else:
        return n*fact(n-1)
n=int(input("n:"))
fact(n)
```

```python
#6. Write a recursive function fun(n) to calculate 2^n.
def two_pow(n):
    if n==1:
        return 2
    else:
        return 2*two_pow(n-1)
n=int(input("n:"))
two_pow(n)
```

```python
#7. Write a recursive function fun(n) to print all even numbers from 1 to n.
def even(n):
    if n==0:
        return
    even(n-1)
    if n%2==0:
        print(n)
n = int(input("n: "))
even(n)
```

```python
#8. Write a recursive function fun(n) to print all odd numbers from n down to 1.
def odd(n):
    if n==0:
        return 
    if n%2!=0:
        print(n)
    odd(n-1)
n = int(input("n: "))
odd(n)
```

```python
#9. Write a recursive function fun(n) to count how many numbers exist from 1 to n.
def count_num(n):
    if n==0:
        return 0
    return 1+count_num(n-1)

n = int(input("n: "))
print(count_num(n))
```

```python
#10. Write a recursive function fun(n) that prints n before the recursive call and prints n again after the recursive call.
def rec(n):
    if n==0:
        return 
    print(n)
    rec(n-1)
    print(n)
n=int(input("n:"))
rec(n)
```

## Level 2- Basic problem solving

```python
#11. Find the sum of all digits of a number using recursion.
def sum_digi(n):
    if n==0:
        return 0
    else:
        return n%10+sum_digi(n//10)

n=int(input("n:"))
sum_digi(n)
```

```python
#12. Count the number of digits in a positive integer using recursion.
def count_num(n):
    if n==0:
        return 0
    else:
        return 1+count_num(n//10)
n=int(input("n:"))
count_num(n)
```

```python
#13. Find the product of all digits of a number using recursion.
def prod(n):
    if n==1:
        return 1
    else:
        return n%10 * prod(n//10)
n=int(input("n:"))
prod(n)
```

```python
#14. Find the largest digit in a number using recursion.
def large(n):
    if n<10:
        return n
    last=n%10
    remaining=large(n//10)
    if last > remaining:
        return last
    else:
        return remaining
n=int(input("n:"))
large(n)
```

```python
#15. Find the smallest digit in a number using recursion.
def small(n):
    if n<10:
        return n
    last=n%10
    remaining=small(n//10)
    if last < remaining:
        return last
    else:
        return remaining
n=int(input("n:"))
small(n)
```

```python
#16. Count how many times digit 5 occurs in a number using recursion.
def count_5(n):
    if n==0:
        return 0
    last=n%10

    if last==5:
        return 1+count_5(n//10)
    else:
        return count_5(n//10)
n=int(input("n:"))
count_5(n)
```

```python
#17. Find the sum of all even digits using recursion.
def sum_even(n):
    if n==0:
        return 0
    digi=n%10
    if digi%2==0:
        return digi+sum_even(n//10)
    else:
        return sum_even(n//10)
n=int(input("n:"))
sum_even(n)
```

```python
#18. Find the sum of all odd digits using recursion.
def sum_odd(n):
    if n==0:
        return 0
    digi=n%10
    if digi%2!=0:
        return digi+sum_odd(n//10)
    else:
        return sum_odd(n//10)
n=int(input("n:"))
sum_odd(n)  
```

```python
#19. Reverse a positive integer using recursion.
def reverse(n,rev=0):
    if n==0:
        return rev
    digi=n%10
    rev=rev*10+digi
    return reverse(n//10,rev)
n=int(input("n:"))
reverse(n)
```

```python
#20. Check whether a number is a palindrome using recursion.
def reverse(n,original=n,rev=0):
    if n==0:
        return original==rev
    digi=n%10
    rev=rev*10+digi
    return reverse(n//10,original,rev)
        
n=int(input("n:"))
reverse(n)
```

## Level 3 Recursion+logic

```python
#21. Find the GCD of two numbers using recursion.
def gcd(a,b):
    if b==0:
        return a
    return gcd(b,a%b)

a=int(input("a:"))
b=int(input("b:"))
print(gcd(a,b))
```

```python
#22. Find the LCM of two numbers using recursion.
def gcd(a,b):
    if b==0:
        return a
    return gcd(b,a%b)
def lcm(a,b):
    return (a*b)/gcd(a,b)
    

a=int(input("a:"))
b=int(input("b:"))
print(lcm(a,b))
```

```python
#23. Find a^b using recursion without using **.
def pow(a,b):
    if b==0:
        return 1
    else:
        return a*pow(a,b-1)
a=int(input("a:"))
b=int(input("b:"))
pow(a,b)
```

```python
#24. Find the sum:1 + 2 + 3 + ... + n using recursion, but your function must return the answer.
def sum_num(n):
    if n==0:
        return 0
    else:
        return n+sum_num(n-1)
n=int(input("n:"))
sum_num(n)
```

```python
#25. Find the sum of: 1² + 2² + 3² + ... + n² using recursion.
def pow_sum(n):
    if n==0:
        return 0
    else:
        return n**2+pow_sum(n-1)
n=int(input("n:"))
pow_sum(n)
```

```python
#26. Find the sum of: 1 + 3 + 5 + ... + (2n-1) using recursion.
def sum(n):
    if n==0:
        return 0
    else:
        return (2*n-1)+sum(n-1)
n=int(input("n:"))
sum(n)
```

```python
#27. Convert a positive decimal number to binary using recursion.
def binary(n):
    if n==0:
        return ""
    return binary(n//2)+str(n%2)
n=int(input("n:"))
binary(n)
```

```python
#28. Find the number of zeros in a positive integer using recursion.
def count_0(n):
    if n==0:
        return 0
    last=n%10
    if last==0:
        return 1+count_0(n//10)
    else:
        return count_0(n//10)

n=int(input("n:"))
count_0(n)
```

```python
#Check whether a number is a power of 2 using recursion.
def power(n):
    if n==1:
        return True
    if n%2!=0:
        return False
    return power(n//2)
n=int(input("n:"))
power(n)
```

```python
#32. Harmonic Series Sum Using Recursion
def harmonic(n):
    if n<2:
        return 1
    else:
        return 1/n+harmonic(n-1)
n=int(input("N:"))
harmonic(n)
```
