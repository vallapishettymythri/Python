## Level-2 Dictionary+loops+logic

```python
#Print subjects having marks greater than 80
def greater(marks):
    for i,j in marks.items():
        if j>80:
            print(i,j)
marks = {'Math': 85, 'Python': 92, 'C': 78, 'Java': 88}
greater(marks)
```

```python
#Count how many values are even
def even(numbers):
    count=0
    for i in numbers.values():
        if i%2==0:
            count+=1
    print(count)
numbers = {'a': 12, 'b': 7, 'c': 20, 'd': 15, 'e': 8}
even(numbers)
```

```python
#Count positive, negative and zero values
def count(numbers):
    pos=0
    neg=0
    zero=0
    for i in numbers.values():
        if i>0:
            pos+=1
        elif i==0:
            zero+=1
        else:
            neg+=1
    print(pos,neg,zero)

numbers = {'a': 10, 'b': -5, 'c': 0, 'd': 8, 'e': -2}
count(numbers)
```

```python
# Find the key having the largest value
def largest(marks):
    large=0
    large_key=""
    for key,value in marks.items():
        if value>large:
            large=value 
            large_key=key
    print(large_key)
marks = {'Math': 85, 'Python': 92, 'C': 78, 'Java': 88}
largest(marks)
            
            
```

```python
#Find the key having the smallest value
def smallest(marks):
    small=list(marks.values())[0]
    small_key=""
    for key,value in marks.items():
        if value<small:
            small=value
            small_key=key
    print(small_key)

marks = {'Math': 85, 'Python': 92, 'C': 78, 'Java': 88}
smallest(marks)
```

```python
#Calculate the sum of all values without sum()
def sums(numbers):
    sum=0
    for i in numbers.values():
        sum+=i
    print(sum)
numbers = {'a': 10, 'b': 20, 'c': 30, 'd': 40}
sums(numbers)
```

```python
#Calculate the average without sum()
def average(marks):
    sum=0
    for i in marks.values():
        sum+=i
    avg=sum/len(marks)
    print(avg)
marks = {'Math': 80, 'Python': 90, 'Java': 70, 'C': 60}
average(marks)    
```

```python
#Create a new dictionary containing only even values
def even(numbers):
    dict={}
    for i,j in numbers.items():
        if j%2==0:
            dict[i]=j
    return dict
numbers = {'a': 10, 'b': 15, 'c': 20, 'd': 7, 'e': 8}
even(numbers)
```

```python
#Replace all negative values with 0
def neg(numbers):
    for i,j in numbers.items():
        if j<0:
            numbers[i]=0
    return numbers
numbers = {'a': 10, 'b': -5, 'c': 8, 'd': -3}
neg(numbers)
```

```python
#Count values greater than a given number
def greater(numbers,limit):
    count=0
    for i in numbers.values():
        if i>limit:
            count+=1
    return count
numbers = {'a': 12, 'b': 25, 'c': 7, 'd': 30, 'e': 18}
limit = 15
greater(numbers,limit)
```

## Level 3 — Functions + Recursion + Dictionary Logic

```python
# Write a function to find the largest value
def largest(marks):
    largest=0
    for i in marks.values():
        if largest<i:
            largest=i
    return largest
marks = {'Math': 85, 'Python': 92, 'Java': 78, 'C': 88}
largest(marks)
```

```python
#Function to return keys whose values are above average
def average(marks):
    sum=0
    for i in marks.values():
        sum+=i
    avg=sum/len(marks)
    for i,j in marks.items():
        if j>avg:
            print(i)
    
marks = {'Math': 85, 'Python': 92, 'Java': 78, 'C': 88}
average(marks)
```

```python
#Remove all keys having odd values
def odd(numbers):
    result={}
    for i,j in numbers.items():
        if j%2==0:
            result[i]=j
    return result
numbers = {'a': 10, 'b': 15, 'c': 20, 'd': 7, 'e': 8}
odd(numbers)
```

```python
#Find the second-largest value
def largest(marks):
    largest=0
    second=0
    for j in marks.values():
        if j>largest:
            second=largest
            largest=j
        if j>second and largest!=j:
            second=j
    return largest,second
marks = {'Math': 85, 'Python': 95, 'Java': 78, 'C': 90}
largest(marks)
            
```

```python
#Find the key corresponding to the second-largest value
def largest(marks):
    largest=0
    second=0
    first_key=""
    second_key=""
    for i,j in marks.items():
        if j>largest:
            second=largest
            largest=j
            first_key=i
        if j>second and largest!=j:
            second=j
            second_key=i
    return first_key,second_key
marks = {'Math': 85, 'Python': 95, 'Java': 78, 'C': 90}
largest(marks)
```

```python
#Group numbers by parity
def parity(lst):
    result={
        "odd":[],
        "even":[]
    }
    for i in lst:
        if i%2==0:
            result["even"].append(i)
        else:
            result["odd"].append(i)
    return result
lst = [1, 2, 3, 4, 5, 6, 7, 8]
parity(lst)
```

```python
#Find words having the same length
def same_length(words):
    result={}
    for i in words:
        length=len(i)
        if length not in result:
            result[length]=[]
        result[length].append(i)
    return result
words = ['cat', 'apple', 'dog', 'ball', 'hi', 'sun']
same_length(words)
```

```python
#Group students according to marks range
def group(students):
    result={
        "90+": [],
        "75-89": [],
        "50-74": [],
        "Below 50": []
    }
    for i,j in students.items():
        if j>=90:
            result["90+"].append(i)

        elif j>=75:
            result["75-89"].append(i)
        elif j>=50:
            result["50-74"].append(i)
        else:
            result["Below 50"].append(i)
    return result
students = {
'Alice': 85,
'Bob': 62,
'Charlie': 91,
'David': 48,
'Eva': 76
}
group(students)
```

```python
#Find common elements between two lists using a dictionary
def common(a,b):
    result={}
    for i in a:
        if i in b:
            result[i]=i
    return result
a = [1, 2, 3, 4, 5]
b = [3, 5, 7, 9]
common(a,b)
```

```python
#Check whether two strings are anagrams
def anagram(str1,str2):
    d1={}
    d2={}
    for i in str1:
        if i not in d1:
            d1[i]=1
        else:
            d1[i]+=1
    for i in str2:
        if i not in d2:
            d2[i]=1
        else:
            d2[i]+=1
    if d1==d2:
        return "Anagram"
    else:
        return "Not anagram"
str1 = "listen"
str2 = "silent"
anagram(str1,str2)
```

```python
#Build a dictionary from a list without using Counter
def dict(lst):
    d={}
    for i in lst:
        if i not in d:
            d[i]=1
        else:
            d[i]+=1
    return d

lst = ['apple', 'banana', 'apple', 'orange', 'banana', 'apple']
dict(lst)
```

```python
#Find the longest word using a dictionary
def longest(words):
    high=0
    answer=""
    for i in words:
        if len(i)>high:
            high=len(i)
            answer=i
    return answer
words = ['cat', 'elephant', 'dog', 'tiger', 'butterfly']
longest(words)
```
