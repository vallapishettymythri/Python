```python
#Simple Calculator
def calculator(a,op,b):
    try:
        if op=="+":
            print(a+b)
        elif op=="-":
            print(a-b)
        elif op=="*":
            print(a*b)
        elif op=="/":
            print(a/b)
        else:
            print("Invalid operator")
    except:
        print("Cannot divide by 0")

try:
    a=int(input())
    op=input()
    b=int(input())
    calculator(a,op,b)
except:
    print("Invalid number")
```

```python
#Division With else
def division(a,b):
    try:
        print(a/b)
    except:
        print("division with zero is not possible")
    else:
        print("Division is successful")

a=int(input())
b=int(input())
division(a,b)
```

```python
#using exception classes or types
def division(a,b):
    try:
        a=input()
        b=input()
        print(int(a)/int(b))
    except ValueError:
        print("not valid num")
    except ZeroDivisionError:
        print("Cant be divided by zero")
    else:
        print("Executed successfully")
division(a,b)
```

```python
#File With finally
def filefinally(file):
    try:
        f1=open(file,"r")
        print(f1.read())
    except:
        print("File not found")
    finally:
        print("File operation completed")

file=input()
        
f1=open(file,"w")
f1.write("python\n")
f1.write("Java")
f1.close()

filefinally(file)        
```

```python
#File Word Counter
#Create a function that opens a file and counts its words.
def wordcounter(file):
    count=0
    try:
        f1=open(file,"r")
        data=f1.read()
        words=data.split()
        for i in words:
            count+=1
        print(count)
    except FileNotFoundError:
        print("File not found")

file=input()

f1=open("data.txt","w")
f1.write("Python is easy\n")
f1.write("I love Python")
f1.close()

wordcounter(file)
```

```python
#File Line Counter
def linecounter(file,data):
    count=0
    try:
        lines=data.splitlines()
        for i in lines:
            count+=1
        return count
    except:
        print("file not found")
    
file=input()
f1=open(file,"r")
data=f1.read()
f1.close()

linecounter(file,data)
```

```python
#Product Quantity
def product(stock,prod,quantity):
    try:
        found==False
        for i,j in stock.items():
            if i==prod:
                 if quantity > j:
                    print("Insufficient stock")
                 else:
                     res = j - quantity
                     print("Order successful")
                     return res
        if found==False:
            raise Exception
                
    except:
        print("product not found")

stock = {
    "Laptop": 5,
    "Mouse": 10,
    "Keyboard": 7
}
prod=input()
quantity=int(input())
product(stock,prod,quantity)
```

```python
#Mini Banking System
def  banking(choice,amount,total):
    try:
        if amount <= 0:
            print("Negative amount")

        elif choice == "deposit":
            total += amount
            print("Deposit successful")
            print("Balance:", total)

        elif choice == "withdraw":
            if amount > total:
                print("Insufficient balance")
            else:
                total -= amount
                print("Withdrawal successful")
                print("Balance:", total)

        elif choice == "check":
            print("Balance:", total)

        else:
            print("Invalid choice")
    except:
        print("Invalid amount")

        
choice=input()
amount=int(input())
total=int(input())
banking(choice,amount,total)
```

```python
def deposit(total, amount):
    try:
        if amount <= 0:
            print("Invalid amount")
        else:
            total += amount
            print("Deposit successful")
            print("Balance:", total)
            return total
    except:
        print("Invalid amount")


def withdraw(total, amount):
    try:
        if amount <= 0:
            print("Invalid amount")
        elif amount > total:
            print("Insufficient balance")
        else:
            total -= amount
            print("Withdrawal successful")
            print("Balance:", total)
            return total
    except:
        print("Invalid amount")


def check_balance(total):
    try:
        print("Balance:", total)
    except:
        print("Error")


total = 10000

choice = input()
amount = int(input())

if choice == "deposit":
    total = deposit(total, amount)

elif choice == "withdraw":
    total = withdraw(total, amount)

elif choice == "check":
    check_balance(total)

else:
    print("Invalid choice")
```
