# Single Inheritance

```python
class Animal:
    def eat(self):
        print("Animal is eating")
class dog(Animal):
    pass

d=dog()
d.eat()
```

```python
class person:
    def __init__(self,name):
        self.name=name
class student(person):
    pass
n=student("Mythri")
print("Name:",n.name)
```

```python
class vehicle:
    def start(self):
        print("Vehicle started")
class car(vehicle):
    def drive(self):
        print("Car is driving")
c=car()
c.start()
c.drive()
```

```python
class employee:
    def __init__(self,name,salary):
        self.name=name
        self.salary=salary
class manager(employee):
    pass
m=manager("Mythri","1c")
print("name:",m.name)
print("salary:",m.salary)
```

```python
class person:
    def __init__(self,name):
        self.name=name
    def display(self):
        print("Name:",self.name)
class student(person):
    def __init__(self,name,course):
        super().__init__(name)
        self.course=course
    def display1(self):
        self.display()
        print("course:",self.course)
s=student("Mythri","AI")
s.display1()
```

```python
class rectangle:
    def __init__(self,length,breadth):
        self.length=length
        self.breadth=breadth
    def area(self):
        print("area:",self.length*self.breadth)
class room(rectangle):
    pass
r=room(10,5)
r.area()
```

```python
class bankaccount:
    def __init__(self,balance):
        self.balance=balance
class savingsaccount(bankaccount):
    def show_balance(self):
        print(self.balance)
b=savingsaccount(15000)
b.show_balance()
```

```python
class employee:
    def __init__(self,name,salary):
        self.name=name
        self.salary=salary
    def display_salary(self):
        print(self.salary)
class manager(employee):
    def increase_salary(self):
        self.salary+=5000
        self.display_salary()
m=manager("aryan",50000)
m.increase_salary()
```

```python
class student:
    def __init__(self,name,marks):
        self.name=name
        self.marks=marks
    def display(self):
        print("name:",self.name)
class result(student):
    def check_result(self):
        self.display()
        if self.marks>=40:
            print("Pass")
        else:
            print("Fail")
s=result("Mythri",85)
s.check_result()
```

```python
class employee:
    def __init__(self,name,salary,performance):
        self.name=name
        self.salary=salary
        self.performance=performance
class manager(employee):
    def evaluate(self):
        print(self.name)
        if self.performance >=90:
            bonus=self.salary/20
            self.salary+=bonus
            print("Excellent",bonus,self.salary)
        elif self.performance >=75:
            bonus=self.salary/15
            self.salary+=bonus
            print("Very Good",bonus,self.salary)
        elif self.performance >=60:
            bonus=self.salary/10
            self.salary+=bonus
            print("Good",bonus)
        else:
            bonus=self.salary/5
            self.salary+=bonus
            print("Needs improvement",bonus,self.salary)
            
e=manager("mythri",60000,85)   
e.evaluate()
```
