### Generator

```python
#Create a generator function numbers() 
#that generates numbers from 1 to 5, one at a time.
def generate():
    i=1
    while i<=5:
        yield i
        i+=1
call=generate()
next(call)
```

```python
#Create a generator even_numbers(n) that generates even numbers from 1 up to n.
def even_num(n):
    i=1
    while i<=n:
        if i%2==0:
            yield i
        i+=1
call=even_num(5)
print(next(call))
```

```python
next(call)
```

```python
#tracing
def test():
    yield 10
    yield 20
    yield 30

x = test()

print(next(x))
print(next(x))
print(next(x))
```

```python
#It should generate the squares from 1 to n.
def square(n):
    for i in range(1,n+1):
        sq=i**2
        yield sq
x=square(5)
print(next(x))
```

```python
next(x)
```

```python
next(x)
```

```python
#Create a generator table(n) that generates 
#the multiplication table of n from 1 to 10.
def multipliction(n):
    i=1
    while i<=10:
        multi=n*i
        yield multi
        i+=1
x=multipliction(5)
print(next(x))
```

```python
next(x)
```

```python
next(x)
```

```python
#Create a generator that generates numbers between 1 and n divisible by 3.
def divisible(n):
    for i in range(1,n+1):
        if i%3==0:
            yield i
x=divisible(20)
next(x)
```

```python
next(x)
```

```python
#Reverse numbers
#Create a generator that generates numbers from n down to 1.
def reverse(n):
    for i in range(n,0,-1):
        yield i
x=reverse(5)
print(next(x))
```

```python
#Create a generator that generates each character of a string one at a time.
def character(s):
    for i in s:
        yield i
x=character("Python")
next(x)
```

```python
next(x)
```

```python
#Filter positive numbers
#Create a generator that receives a list and generates only positive numbers.
def positive(lst):
    for i in lst:
        if i>0:
            yield i

lst=[-3, 5, -1, 8, 0, 2]
x=positive(lst)
next(x)
```

```python
next(x)
```

```python
next(x)
```

```python
#Fibonacci generator
#Create a generator that generates the first n Fibonacci numbers.
def fibo(n):
    a=0
    b=1

    for i in  range(n):
        yield a
        c=a+b
        a=b
        b=c
x=fibo(7)
for value in x:
    print(value)
```
